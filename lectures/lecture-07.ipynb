### 1. The for Loop and range() Function
A for loop is typically used  to repeat a set of statements a specific number of times.

```python
for i in range(10):
  print(i)
```

```python
times_to_repeat = range(10)

for i in times_to_repeat:
  print(i)
```

### 2. The range() Function and for Loops
The range() function used with for loops to repeat a block of code a specific number of times.
- range(upper)
- range(lower, upper)
- range(lower, upper, increment/decrement)

```python
for i in 10:
    print(i, end=" ")
```

```python
for i in range(5):
    print(i, end=" ")
print("end")
```

```python
print("range(1, 5):")
for i in range(1, 5):
    print(i, end=" ")
print("\n")
```

```python
print("range(2, 10, 2):")
for i in range(2, 10, 2):
    print(i, end=" ")
```

More for Loop Examples

```python
print("range(10, 2, -2):")
for i in range(10, 2, -2):
    print(i, end=" ")
```

```python
print("range(10, 2, -2):")
for i in range(100, -100, -10):
    print(i, end=" ")
```

### 2. Loop Control: break and continue
These statements allow you to change the flow of a loop based on conditions:

- break: Jumps out of the loop immediately.

- continue: Jumps back to the beginning of the loop for the next iteration.

```python
while True:
    miles = float(input("Enter miles: "))
    if miles <= 0:
        print("Must be greater than zero. Try again.")
        continue

    choice = input("Enter another? (y/n): ")
    if choice.lower() != 'y':
        break  # Exits the loop entirely
```

### 3. The Walrus Operator (:=)
formally known as the assignment expression, allows you to assign a value to a variable within another expression.

```python
# Without the walrus operator
print("Enter -1 to quit.")
while True:
    score = input("Enter a score: ")
    if score == "-1":
        break
    print(f"You entered {score}.")
```

```python
# Using the Walrus Operator
print("Enter -1 to quit.")

while (score := input("Enter a score: ")) != "-1": # Assigns input to 'score' AND checks the condition in one line
    print(f"You entered {score}.")

print("Bye!")
```

```python
def calculate_miles_per_gallon(miles_driven, gallons):
    mpg = miles_driven / gallons
    mpg = round(mpg, 1)
    return mpg
```

```python
def calculate_future_value(monthly_investment, yearly_interest,
                           years):
    # convert yearly values to monthly values
    monthly_interest_rate = yearly_interest / 12 / 100
    months = years * 12

    # calculate future value
    future_value = 0.0
    for i in range(months):
        future_value += monthly_investment
        monthly_interest = future_value * monthly_interest_rate
        future_value += monthly_interest
    return future_value
```

```python
def main():
    choice = "y"

    while choice.lower() == "y":
        # get input from the user
        monthly_investment = float(input(
            "Enter monthly investment:\t"))
        yearly_interest_rate = float(input(
            "Enter yearly interest rate:\t"))
        years = int(input("Enter number of years:\t\t"))

        # get and display future value
        future_value = calculate_future_value(
            monthly_investment, yearly_interest_rate, years)

        print(f"Future value:\t\t\t{round(future_value, 2)}")
        print()

        # see if the user wants to continue
        choice = input("Continue? (y/n): ")
        print()

    print("Bye!")
```
