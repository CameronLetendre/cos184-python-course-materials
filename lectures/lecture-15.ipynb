### 1. File Basics and the open() Function (The File Handler Example)

open(): A built-in function used to open a file and return a file object.


'r': The "Read" mode, used to open a file for reading (default).


'w': The "Write" mode, used to open a file for writing; it overwrites any existing data in the file.


'a': The "Append" mode, used to open a file and add new data to the end of its existing content.


close(): A method used to manually close a file object to free up system resources.

```python
# Manually opening and closing a file
outfile = open("test.txt", "w")
outfile.write("Test")
outfile.close()
```

### 2. The with Statement

with: A statement that provides a context manager for file I/O. It automatically closes the file object once the block of code is finished, even if an exception occurs.

```python
with open("test.txt", "w") as outfile:
    outfile.write("Hello World")

with open("test.txt", "r") as infile:
    content = infile.readline()
    print(content)
```

Using a variable name for file path

```python
filename = "test.txt"

with open(filename, "w") as outfile:
    outfile.write("Hello World")

with open(filename, "r") as infile:
    content = infile.readline()
    print(content)
```

### 3. File Writing Methods

write(str): A method of a file object that writes a single string to the file. It does not automatically include a newline character (\n), so one must often be added manually.

```python
members = ["John Cleese", "Eric Idle"]

with open("members.txt", "w") as file:
    for member in members:
        file.write(f"{member}\n")  # Manually adding newlines
```

```python
with open("members.txt", "a") as file:
    file.write("John Cheese\n")
```

Using write again overwrites the file

```python
members = ["John Cleese", "Eric Idle"]

with open("members.txt", "w") as file:
    for member in members:
        file.write(f"{member}\n")  # Manually adding newlines
```

### 4. File Reading Methods
read(): Reads the entire content of a file into a single

string.readline(): Reads a single line from the file, moving the file pointer to the next

line.readlines(): Reads all lines from a file and stores them as a list of strings.

for line in file: An efficient way to iterate through a file line by line without loading the entire file into memory at once

```python
# Reading the entire file as one string
with open("members.txt") as file:
    full_string = file.read()
    print(full_string)
```

```python
# Reading the entire file line by line
with open("members.txt") as file:
    line = file.readline()
    print(line)
    line = file.readline()
    print(line)

```

```python
# Reading the entire file into a list
with open("members.txt") as file:
    line = file.readlines()
    print(line)

```

```python
# Reading lines into a clean list iteratively
processed_list = []

with open("members.txt") as file:
    for line in file:
        processed_list.append(line)

print(f"Clean List: {processed_list}")
```

### 5. Working with Numerical Data (The Year Tracker Example)

Data Conversion: Since text files only store strings, numeric data must be converted to strings (e.g., via f-strings) when writing and converted back to numbers (using int() or float()) when reading.

```python
years = [1975, 1979, 1983]

# Writing numbers to a text file
with open("years.txt", "w") as years_file:
    for year in years:
        years_file.write(f"{year}\n") # Convert int to str
```

```python
# Reading numbers back into integer format
restored_years = []
with open("years.txt") as file:
    for line in file:
        restored_years.append(int(line)) # Convert str back to int

print(f"Restored Years: {restored_years}")
```

### Movie list program

```python
FILENAME = "../data/movies.txt"

def write_movies(movies):
    with open(FILENAME, "w") as file:
        for movie in movies:
            file.write(f"{movie}\n")

def read_movies():
    movies = []
    with open(FILENAME) as file:
        for line in file:
            line = line.replace("\n", "")
            movies.append(line)
    return movies

def list_movies(movies):
    for i, movie in enumerate(movies, start=1):
        print(f"{i}. {movie}")
    print()

def add_movie(movies):
    movie = input("Movie: ")
    movies.append(movie)
    write_movies(movies)
    print(f"{movie} was added.\n")

def delete_movie(movies):
    index = int(input("Number: "))
    if index < 1 or index > len(movies):
        print("Invalid movie number.\n")
    else:
        movie = movies.pop(index - 1)
        write_movies(movies)
        print(f"{movie} was deleted.\n")

def display_menu():
    print("The Movie List program")
    print()
    print("COMMAND MENU")
    print("list - List all movies")
    print("add -  Add a movie")
    print("del -  Delete a movie")
    print("exit - Exit program")
    print()

def main():
    display_menu()
    movies = read_movies()
    while True:
        command = input("Command: ")
        if command.lower() == "list":
            list_movies(movies)
        elif command.lower() == "add":
            add_movie(movies)
        elif command.lower() == "del":
            delete_movie(movies)
        elif command.lower() == "exit":
            print("Bye!")
            break
        else:
            print("Not a valid command. ",
                  "Please try again.")

if __name__ == "__main__":
    main()
```

```python
import csv

movies = [
    ["Inception", "Sci-Fi", 2010],
    ["The Godfather", "Crime", 1972],
    ["Interstellar", "Sci-Fi", 2014]
]

# Use "w" for text mode and newline="" to prevent extra blank rows
with open("movies.tsv", "w") as file:
    writer = csv.writer(file, delimiter="\t")

    writer.writerows(movies)

print("File saved successfully as a tab-separated file.")
```

changing the delimiter for reading

```python
import csv

recovered_movies = []

with open("movies.tsv", "r") as file:
    reader = csv.reader(file, delimiter="\t")

    for row in reader:
        recovered_movies.append(row)

print(f"Data: {recovered_movies}")
```

quotechar

```python
import csv

tricky_movies = [
    ["Everything, Everywhere, All at Once", "Sci-Fi"],
    ["The Secret Life of Pets", "Comedy"],
    ["Interstellar", "Movie"]
]

with open("tricky_movies.csv", "w", newline="", encoding="utf-8") as file:

    # quotechar='"' is the default
    # quoting=csv.QUOTE_MINIMAL tells it to only use quotes when necessary
    writer = csv.writer(file, delimiter=",", quotechar='"', quoting=csv.QUOTE_MINIMAL)

    writer.writerow(["Title", "Genre"])
    writer.writerows(tricky_movies)

with open("tricky_movies.csv", "r") as file:
    print(file.read())
```

### Floor and Ceil functions

```python
result = m.floor(12.545) # 12
print(f"{result}")
result = m.ceil(12.545) # 13
print(f"{result}")
result = m.floor(-3.432) # -4
print(f"{result}")
result = m.ceil(-3.432) # -3
print(f"{result}")
```

The ceil() function with decimal places


```python
print(m.ceil(2.0083)) # 3​
print(m.ceil(2.0083 * 10) / 10) # 2.1​
print(m.ceil(2.0083 * 100) / 100) # 2.01​
```

The floor() function with decimal places

```python
result = m.floor(2.0083) # 2​
print(f"{result}")

result = m.floor(2.0083 * 10) / 10 # 2.0​
print(f"{result}")

result = m.floor(2.0083 * 1000) / 1000 # 2.008​
print(f"{result}")
```
