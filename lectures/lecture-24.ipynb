### 1. The Dictionary Update Operator (`|=`)
In Python 3.9 and later, you can use the update operator (`|=`) to merge the contents of one dictionary into another.
* The syntax is `dictionary1 |= dictionary2`.
* If a key from the second dictionary already exists in the first, its value is replaced.
* If the key does not exist, it is added as a new key/value pair.

```python
fruits_1 = {"banana": 1099, "pear": 34, "grape": 67}
fruits_2 = {"apple": 10, "durian": 34, "kiwi": 67}

fruits_1 |= fruits_2
print(fruits_1)
print(fruits_2)
```

```python
person_1 = {"name": "frank", "email": "runner@aol.com", "number": "207-666-7777"}
person_2 = {"email": "legohero@lego.com", "Postal code": "000-111-1390"}

person_1 |= person_2
print(person_1)
print(person_2)
```

### 2. Dictionaries Containing Other Dictionaries
A dictionary can contain other dictionaries as its values, which is extremely useful for storing complex, structured data like a contact book.
* You can access the data inside an embedded dictionary by chaining the brackets: `dictionary_name[key1][key2]`.
* If you try to access a key that does not exist in the inner dictionary, Python will throw a `KeyError`.

```python
contacts = {
    "Joel": {"city": "San Francisco", "phone": "555-555-1111"},
    "Anne": {"city": "Fresno", "phone": "555-555-2222"}
}

print(contacts)
```

```python
contacts = {
    "Joel": {"city": "San Francisco", "phone": "555-555-1111"},
    "Anne": {"city": "Fresno", "phone": "555-555-2222"}
}


joel_city = contacts["Joel"]["city"]
anne_phone = contacts["Anne"]["phone"]

print(f"Joel's city: {joel_city}")
print(f"Anne's phone: {anne_phone}")
```

```python
contacts = {
    "Joel": {"city": "San Francisco", "phone": "555-555-1111"},
    "Anne": {"city": "Fresno", "phone": "555-555-2222"}
}

joel_city = contacts["Joel"]["email"]

print(f"Joel's email: {joel_city}")
```

### 3. Safely Accessing Embedded Dictionaries
To avoid `KeyError` or `AttributeError` exceptions when dealing with nested dictionaries, you should safely check if the keys exist.
* You can check if a key exists within another key using the `in` keyword: `if inner_key in dictionary[outer_key]:`.
* You can also safely chain `.get()` methods. By supplying an empty dictionary `{}` as the default value for the first `.get()`, you ensure the second `.get()` won't crash if the outer key is missing: `dictionary.get(outer_key, {}).get(inner_key)`.

```python
contacts = {
    "Anne": {"city": "Fresno", "phone": "555-555-2222"}
}

# Method 1: Using the 'in' keyword to check safely
key = "phone"

if key in contacts["Anne"]:
    email = contacts["Anne"][key]
    print(email)
else:
    print("Sorry, there is no email address for this contact.")
```

```python
contacts = {
    "Joel": {"city": "San Francisco", "phone": "555-555-1111"},
    "Anne": {"city": "Fresno", "phone": "555-555-2222"}
}

if "phone" in contacts["Joel"]:
  joel_email = contacts["Joel"]["phone"]

  print(f"Joel's email: {joel_email}")
else:
    print(f"Joel's email not provided")
```

### The Book Catalog program

```python
def show_book(book_catalog):

    for title in book_catalog:
        print(f"Title: {title}")
    else:
        print(f"Sorry, {title} doesn't exist in the catalog.")
```

```python
def add_edit_book(book_catalog, mode):

    title = input("Title: ")

    if mode == "add" and title in book_catalog:
        print(f"{title} already exists in the catalog.")
        response = input (
        "Would you like to edit it? (y/n): ").lower()
        if(response != "y"):
            return
    elif mode == "edit" and title not in book_catalog:
        print(title + " doesn't exist in the catalog.")
        response = input(
        "Would you like to add it? (y/n): ").lower()
        if (response != "y"):
            return

    # Get remaining book data and create a dictionary for the data

    author = input("Author name: ")

    pubyear = input("Publication year: ")

    book = {title: {"author": author, "pubyear": pubyear}}

    book_catalog |= book  #TODO Add the book data to the catalog using the update operator
```

```python
def delete_book(book_catalog):
    title = input("Title: ")

    if title in book_catalog:
        book_catalog.pop(title)             #TODO delete the title in the catalog
        print(f"{title} removed from catalog.")
    else:
        print(f"{title} doesn't exist in the catalog.")

def display_menu():
    print("The Book Catalog program")
    print()
    print("COMMAND MENU")
    print("show - Show book info")
    print("add - Add book")
    print("edit - Edit book")
    print("del - Delete book")
    print("exit - Exit program")
```

```python
def main():
    book_catalog = {
        "Moby Dick": {"author": "Herman Melville", "pubyear": "1851"},
        "The Hobbit": {"author": "J. R. R. Tolkien", "pubyear": "1937"},
        "Slaughterhouse Five": {"author": "Kurt Vonnegut", "pubyear": "1969"}
    }

    display_menu()

    while True:
        print()
        command = input("Command: ").lower()

        if command == "show":
            show_book(book_catalog)
        elif command == "add":
            add_edit_book(book_catalog, mode="add")
        elif command == "edit":
            add_edit_book(book_catalog, mode="edit")
        elif command == "del":
            delete_book(book_catalog)
        elif command == "exit":
            print("Bye!")
            break
        else:
            print("Unknown command. Please try again.")

if __name__ == "__main__":
    main()
```

### sys.exit() for ending a program

```python
import sys

filename = input("Enter filename: ")

movies = []

try:
    with open(filename) as file:
        for line in file:
            line = line.replace("\n", "")
            movies.append(line)
except FileNotFoundError as e:
    print("FileNotFoundError:", e)
    sys.exit() # Since we are using a virtual enviroment this will also throw an error
except OSError as e:
    print("OSError:", e)
    sys.exit()
except Exception as e:
    print(type(e), e)
    sys.exit()
```

### Complete Try statement syntax

```python
import csv
def read_movies(filename):
    try:
        with open(filename, newline="") as file:
            movies = []
            reader = csv.reader(file)
            for row in reader:
                movies.append(row)
            return movies
    except Exception as e:
        print(e)
```

```python
import csv
def read_movies(filename):
    try:
        file = open(filename, newline="")
        try:
            movies = []
            reader = csv.reader(file)
            for row in reader:
                movies.append(row)
            return movies
        except Exception as e:
            print(type(e), e)
        finally:
            print("In finally")
            file.close()
    except FileNotFoundError as e:
        print(e)
```
