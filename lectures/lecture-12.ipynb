### 1. Mutable vs. Immutable Types & Arguments
- Immutable Types: str, int, float, and bool are immutable, meaning their objects cannot be changed in place.

- Mutable Types: list is a mutable type, meaning its contents can be altered after creation.

When passing an immutable object to a function, you must return the new object if you want to save changes.

Immutable

```python
def double_the_number(value):
    value = value * 2  # New int object created
    return value       # Must return the new object

value1 = 25
value2 = double_the_number(value1)
```

Mutable

```python
def add_to_list(my_list, item):
    my_list.append(item)  # Modifies the original list object

inventory = ["staff", "hat", "bread"]
add_to_list(inventory, "robe")
# No return needed; inventory is now ["staff", "hat", "bread", "robe"]
```

### 2. Processing Lists with enumerate()
The built-in enumerate() function is used to get an automatic counter value when looping through list items.

You can pass an optional start argument to begin counting from a specific number (like 1 instead of the default 0).

```python
inventory = ["staff", "hat", "bread", "potion"]

for i, item in enumerate(inventory, start=1):
    print(f"{i}. {item}")
```

### 3. Processing Multiple Lists in Parallel with zip()
The zip() function allows you to iterate through two or more distinct lists simultaneously.

```python
inventory = ["staff", "hat", "bread", "potion"]
prices = [27.99, 10.99, 5.99, 19.99]

for item, price in zip(inventory, prices):
    print(f"{item} (${price})")
```

### 4. Basic List Modification Commands
Programs often use a while True loop and conditional statements to create a command menu for users to manage lists dynamically.

Items can be added using the .append() method and removed by index using .pop()

```python
def display_menu():
    print("COMMAND MENU")
    print("list - List all movies")
    print("add  - Add a movie")
    print("del  - Delete a movie")
    print("exit - Exit program")
    print()

def list(movie_list):
    for i, movie in enumerate(movie_list, start=1):
        print(f"{i}. {movie}")
    print()

def add(movie_list):
    movie = input("Name: ")
    movie_list.append(movie)
    print(f"{movie} was added.\n")

def delete(movie_list):
    number = int(input("Number: "))
    if number < 1 or number > len(movie_list):
        print("Invalid movie number.\n")
    else:
        movie = movie_list.pop(number-1)
        print(f"{movie} was deleted.\n")

def main():
    movie_list = ["Monty Python and the Holy Grail",
                  "On the Waterfront",
                  "Cat on a Hot Tin Roof"]

    display_menu()
    while True:
        command = input("Command: ")
        if command.lower() == "list":
            list(movie_list)
        elif command.lower() == "add":
            add(movie_list)
        elif command.lower() == "del":
            delete(movie_list)
        elif command.lower() == "exit":
            break
        else:
            print("Not a valid command. ",
                  "Please try again.\n")

    print("Bye!")

if __name__ == "__main__":
    main()
```

### 6. Accessing and Iterating Through a List of Lists
Elements in a 2D list are accessed using two sets of square brackets, where the first represents the row index and the second represents the column index.

You can print a 2D list's contents clearly by nesting two for loops.

```python
first_movie_name = movies[0][0]
first_movie_price = movies[0][2]

for movie in movies:          # Outer loop accesses each row
    for item in movie:        # Inner loop accesses columns in that row
        print(item, end=" | ")
    print()
```

### Sort with mixed cases

```python
inventory = ["staff", "hat", "shoes", "bread", "potion", ]

stuff = [inventory[0]]

for i in range(0, len(inventory)): # Loop through inventory list
  for j in range(0, len(stuff)): # Loop through current stuff list
    if inventory[i] < stuff[j]: # if current inventory item is less than current stuff list
      stuff.insert(j, inventory[i])
      break

print(stuff)
```

### shallow copy

```python
import copy

original = [[1, 2, 3], [4, 5, 6]]
shallow = copy.copy(original)

shallow.append([7,8,9])

print(original)
print(shallow)
```

```python
import copy

original = [[1, 2, 3], [4, 5, 6]]
shallow = copy.copy(original)

shallow[0][0] = "X"

shallow.append([7,8,9])

print(original)
print(shallow)
```

DeepCopy


```python
import copy

original = [[1, 2, 3], [4, 5, 6]]
deep = copy.deepcopy(original)

deep[0][0] = "X"

print(original)
print(deep)
```

# Slicing Lists

```python
numbers = [52, 54, 56, 58, 60, 62]

print(numbers[0:2]) # [52, 54]
print(numbers[:2]) # [52, 54]

print(numbers[4:]) # [60, 62]
```

```python
numbers = [52, 54, 56, 58, 60, 62]

print(numbers[0:4:2]) # [52, 56]​
print(numbers[::-1]) # [62, 60, 58, 56, 54, 52]
```
