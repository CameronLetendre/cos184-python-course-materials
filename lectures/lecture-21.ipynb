### 1: Parsing Strings into Datetime Objects
The strptime() method of the datetime class allows you to create datetime objects from strings using format codes. Common format codes include %m for the month as a number, %d for the day of the month, and %Y for a 4-digit year

```python
from datetime import datetime

date_str = "10/31/1988"
birth_date = datetime.strptime(date_str, "%m/%d/%Y")

print("Parsed datetime object:", birth_date)
```

```python
from datetime import datetime

date_str = "10-31-1988"
birth_date = datetime.strptime(date_str, "%m-%d-%Y")

print(birth_date)
```

```python
from datetime import datetime

date_str = "10-31-1988 12:30"
birth_date = datetime.strptime(date_str, "%m-%d-%Y %H:%M")

print(birth_date)
```

```python
date_str = "10/31/1988"
birth_date = datetime.strptime(date_str, "%m%d%y")

print("Parsed datetime object:", birth_date)
```

### 2: Formatting Datetime Outputs
You can also format dates for display using these formatting codes directly within f-strings. You can use codes like %B for the full month name, %A for the full weekday name, %I for the hour in a 12-hour format, and %p for the AM/PM specifier

```python
from datetime import datetime

halloween = datetime(1988, 10, 31, 22, 48)

print(f"Standard Format: {halloween:%Y-%m-%d}")
```

```python
halloween = datetime(1988, 10, 31)
print(f"Expanded Format: {halloween:%B %d, %Y %H}")
```

```python
halloween = datetime(1988, 10, 31, 22, 48)

print(f"12-Hour Time Format: {halloween:%B %d, %I:%M %p}")
```

### 3: Creating and Applying Timedeltas
A timedelta object represents a specific span of time. The constructor allows you to build this span using weeks, days, hours, minutes, seconds, milliseconds, and microseconds. You can add or subtract these spans directly from current dates and times

```python
from datetime import date, datetime, timedelta

# Adding and subtracting a span of time
three_weeks_from_today = date.today() + timedelta(weeks=3)

print("Three weeks from today:", three_weeks_from_today)
```

```python
three_hours_ago = timedelta(hours=2) + timedelta(hours=2)
print(datetime.now())

print("Three hours ago:", three_hours_ago)
```

### 4: Calculating Differences Between Dates
Subtracting one datetime object from another automatically generates a timedelta object representing the span between the two dates. You can then extract data from this span using the days, seconds, and microseconds attributes, or by calling the total_seconds() method

```python
from datetime import datetime

# Getting the time span between two dates
halloween = datetime(2021, 10, 31)
time_span = halloween - datetime.now()
print(time_span)

# Extracting properties from the timedelta object
days_difference = time_span.seconds * 60

print(f"Days between now and Halloween 2021: {abs(days_difference)}")
```

```python
from datetime import datetime

halloween = datetime(2021, 10, 31)
time_span = halloween - datetime.now()

total_seconds_difference = time_span.total_seconds()
print(f"Total seconds difference: {abs(total_seconds_difference)}")
```

### Finding the number of days until a specific day

```python
from datetime import date

today = date(2026, 12, 20)
halloween = date(today.year, 10, 31)


if today > halloween:
    next_year = today.year + 1
    halloween = date(next_year, 10, 31)


days_until = (halloween - today).days
print(f"{days_until} day(s) until Halloween.")
```

### Comparing two datetime objects

```python
from datetime import datetime, timedelta, time

meeting_start = datetime(2026, 4, 14, 11, 00)
meeting_end = meeting_start + timedelta(hours=1, minutes=15)
now = datetime.now() - timedelta(hours=5)

print(f"Meeting Start: {meeting_start: %H:%M}")
print(f"Meeting End: {meeting_end: %H:%M}")
print(f"Current Time: {now: %H:%M}\n")

if now > meeting_start and now < meeting_end:
    print("This meeting is happening now.")

elif now < meeting_start:
    print("This meeting is coming up.")

elif now > meeting_end:
    print("This meeting already took place.")
```
