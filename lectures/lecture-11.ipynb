### 1. append(), insert(), and remove() Methods
These methods are used to modify the contents of an existing list:

- append(item): Adds a new item to the very end of the list.
- insert(index, item): Adds an item at a specific index, shifting subsequent items to the right.
- remove(item): Removes the first occurrence of the specified item from the list.

```python
inventory = ["staff", "hat", "shoes", "potion"]
inventory.append("bread")        # Adds "bread" to the end
inventory.insert(3, "robe")      # Inserts "robe" at index 3
inventory.remove("shoes")       # Removes "shoes" from the list
```

### 2. pop() Method
The pop() method removes an item from the list and returns that item so you can use it or store it in a variable:
- pop(): Removes and returns the last item in the list if no index is provided.
- pop(index): Removes and returns the item at the specific index you provide

```python
inventory = ["staff", "hat", "robe", "bread"]

print(inventory, "\n")
last_item = inventory.pop()
print(f"list: {inventory}\npopped item: {last_item}\n")

middle_item = inventory.pop(1)
print(f"list: {inventory}\npopped item: {middle_item}")
```

### 3. index() + pop()
You can combine these two methods to find an item's location and then remove it:


index(item): Searches for a specific item and returns its index (position).


Combined use: First, find the index of the item you want, then pass that index to pop() to remove it.

```python
inventory = ["staff", "hat", "robe", "bread"]
i = inventory.index("hat")

print(f"Index of Hat:\t{i}")
```

```python
inventory = ["staff", "hat", "robe", "bread"]
print(inventory, "\n")
i = inventory.index("hat")

item = inventory.pop(i)
print(f"{inventory}\npopped item: {item}")
```

### 4. len() Function
The len() function is a built-in Python function used to determine the number of items currently in a list. It is frequently used in while loops to ensure the loop doesn't exceed the list's boundaries

```python
scores = [70, 80, 90, 100]
count = len(scores)

print(count)
print(len(scores))
```

### 5. Using the in Keyword (if in)
The in keyword is used to check for the existence of an item within a list. This is often used inside an if statement to prevent errors (like trying to remove an item that doesn't exist)

```python
inventory = ["staff", "hat", "bread", "potion"]
item = "bread"

if item in inventory:
    print(item in inventory)
```

### 6. Looping Through a List (for in)
The for loop is the standard way to iterate through every item in a list. The loop assigns each item in the list to a temporary variable (often called item) one by one until it reaches the end.

```python
inventory = ["staff", "hat", "shoes"]
for item in inventory:
    print(item)
```

```python
scores = [70, 80, 90, 100]
total = 0
for score in scores:
    total += score
print(total)
```

# enumerate() and zip() functions

counter

```python
inventory = ["staff", "hat", "bread", "potion"]

i = 1
for item in inventory:
  print(f"{i}. {item}")
  i += 1

```

range()

```python
inventory = ["staff", "hat", "bread", "potion"]

for i in range(len(inventory)):
  item = inventory[i]
  print(f"{i + 1}. {item}")
```

enumerate()

```python
inventory = ["staff", "hat", "bread", "potion"]

for i, item in enumerate(inventory):
  print(f"{i}. {item}")
```

```python
inventory = ["staff", "hat", "bread", "potion"]

for i, item in enumerate(inventory, start=1):
  print(f"{i}. {item}")
```

### Processing two Lists in parallel

```python
inventory = ["staff", "hat", "bread", "potion"]
prices = [27.99, 10.99, 5.99, 19.99]

for item, price in zip(inventory, prices):
  print(f"{item} (${price})")
```

# Movie List Program

```python
def display_menu():
    print("COMMAND MENU")
    print("list - List all movies")
    print("add  - Add a movie")
    print("del  - Delete a movie")
    print("exit - Exit program")
    print()

def list(movie_list):
    for i, movie in enumerate(movie_list, start=1):
        print(f"{i}. {movie}")
    print()

def add(movie_list):
    movie = input("Name: ")
    movie_list.append(movie)
    print(f"{movie} was added.\n")

def delete(movie_list):
    number = int(input("Number: "))
    if number < 1 or number > len(movie_list):
        print("Invalid movie number.\n")
    else:
        movie = movie_list.pop(number-1)
        print(f"{movie} was deleted.\n")

def main():
    movie_list = ["Monty Python and the Holy Grail",
                  "On the Waterfront",
                  "Cat on a Hot Tin Roof"]

    display_menu()
    while True:
        command = input("Command: ")
        if command.lower() == "list":
            list(movie_list)
        elif command.lower() == "add":
            add(movie_list)
        elif command.lower() == "del":
            delete(movie_list)
        elif command.lower() == "exit":
            break
        else:
            print("Not a valid command. ",
                  "Please try again.\n")

    print("Bye!")

if __name__ == "__main__":
    main()
```
