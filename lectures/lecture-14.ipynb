### 1. The functools reduce() Method
reduce(): A function from the functools module that applies a specified function to the items of a list (or iterable) cumulatively, from left to right, to reduce the sequence to a single value. It also accepts an optional start value. Note that using start= as a keyword argument will result in a TypeError

```python
import functools

numlist = [1, 2, 3, 4, 5, 6]

def add_square(total, current):
    return total + (current * current)

total = functools.reduce(add_square, numlist)

print(f"Total: {total}")
```

```python
import functools

numlist = [1, 2, 3, 4, 5, 6]

def add_square(total, current):
    return total + (current * current)

total_with_start = functools.reduce(add_square, numlist, 10)

print(f"Total (Start 10): {total_with_start}")
```


### 1. List Comprehensions

List Comprehension (Basic): Creates a new list by applying an expression to each item in an existing list, replacing the need for a standard for loop.


List Comprehension (Filtering): Allows you to filter the items processed by adding an if conditional expression at the end.


List Comprehension (Functions): Allows calling custom functions within the expression or condition of the comprehension.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_squares = [n * n for n in numbers]

print(f"Even Squares: {even_squares}")
```

```python
numbers = [1, 2, 3, 4, 5, 6]

even_squares = [n for n in numbers if n % 2 == 0]

print(f"Even Squares: {even_squares}")
```

```python
numbers = [1, 2, 3, 4, 5, 6]

even_squares = [n * n for n in numbers if n % 2 == 0]

print(f"Even Squares: {even_squares}")
```

```python
numbers = [1, 2, 3, 4, 5, 6]
inventory = ["shoes", "cape", "bread", "hat"]

combined = []

for n in numbers:
  for i in inventory:
    combined.append([i, n])

print(combined)
```

### 2. Tuples (Creation & Access)
Tuple Creation: Created using parentheses () instead of square brackets []. Tuples are immutable, meaning their items cannot be modified or reassigned once created.


Tuple Access: Items in a tuple can be accessed or sliced using index values in square brackets, just like a list

```python
herbs = ("lavender", "pokeroot", "chamomile", "valerian")

# Accessing and slicing tuple items
first_herb = herbs[0]
sliced_herbs = herbs[1:4]

print(f"First herb: {first_herb}")
print(f"Sliced: {sliced_herbs}")
```

```python
herbs = ("lavender", "pokeroot", "chamomile", "valerian")

herbs[0] = "lavender"
```

### 3. Tuple Unpacking & Functions

Tuple Unpacking: Allows you to assign the individual items inside a tuple to separate variables in a single, clean statement.

Returning Tuples: A function can return multiple values separated by commas, automatically packing them into a tuple. The calling code can then immediately unpack the returned tuple.

```python
tuple_values = (1, 2, 3)

# Unpacking the tuple into variables
a, b, c = tuple_values

print(f"Values: {a}, {b}, {c}")
```

```python
tuple_values = (1, 2, 3)

# Unpacking the tuple into variables
a, b = tuple_values

print(f"Values: {a}, {b}, {c}")
```

```python
def coordinates():
  return 1,2,3

x, y, z = coordinates()

print(f"Values: {x}, {y}, {z}")
```

### 4. Mathematical List/Tuple Functions (Number Cruncher Example)

min(): Returns the lowest value found in a list or tuple.


max(): Returns the highest value found in a list or tuple.


round(): Rounds a calculated number to a specified precision (commonly used when calculating averages).

```python
data_list = [4, 6, 19, 22, 26, 29, 29, 39, 42, 45, 47]

minimum = min(data_list)
maximum = max(data_list)
average = round(sum(data_list) / len(data_list))

print(f"Min: {minimum}")
print(f"Max: {maximum}")
print(f"Average: {average}")
```

Movie list program

```python
FILENAME = "../data/movies.txt"

def write_movies(movies):
    with open(FILENAME, "w") as file:
        for movie in movies:
            file.write(f"{movie}\n")

def read_movies():
    movies = []
    with open(FILENAME) as file:
        for line in file:
            line = line.replace("\n", "")
            movies.append(line)
    return movies

def list_movies(movies):
    for i, movie in enumerate(movies, start=1):
        print(f"{i}. {movie}")
    print()

def add_movie(movies):
    movie = input("Movie: ")
    movies.append(movie)
    write_movies(movies)
    print(f"{movie} was added.\n")

def delete_movie(movies):
    index = int(input("Number: "))
    if index < 1 or index > len(movies):
        print("Invalid movie number.\n")
    else:
        movie = movies.pop(index - 1)
        write_movies(movies)
        print(f"{movie} was deleted.\n")

def display_menu():
    print("The Movie List program")
    print()
    print("COMMAND MENU")
    print("list - List all movies")
    print("add -  Add a movie")
    print("del -  Delete a movie")
    print("exit - Exit program")
    print()

def main():
    display_menu()
    movies = read_movies()
    while True:
        command = input("Command: ")
        if command.lower() == "list":
            list_movies(movies)
        elif command.lower() == "add":
            add_movie(movies)
        elif command.lower() == "del":
            delete_movie(movies)
        elif command.lower() == "exit":
            print("Bye!")
            break
        else:
            print("Not a valid command. ",
                  "Please try again.")

if __name__ == "__main__":
    main()
```
