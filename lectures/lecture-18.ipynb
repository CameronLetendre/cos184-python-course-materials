### 1. Creating Decimal Objects
Exact Precision: Decimal objects take up more memory and processing time than standard floats because they programmatically store the value, but they prevent hidden floating-point math errors.

Passing Strings: To maintain this exact precision, you must pass the numbers into the Decimal() object as strings (e.g., Decimal("100.10")).

```python
import decimal

num_1 = decimal.Decimal("1.00")
num_2 = decimal.Decimal("2.00")


print(type(num_2))
print(num_1 + num_2)

```

### 3. Breaking Down Decimal Parts (as_tuple())
Here you can see exactly how the Decimal object programmatically stores a value. Using the .as_tuple() method, which breaks the number into three distinct parts.

Sign: A flag where 0 is positive and 1 is negative.

Digits: A tuple holding the raw numbers, completely ignoring the decimal point.

Exponent: A negative number dictating how many places to move the decimal to the left.

```python
from decimal import Decimal

my_number = Decimal('-3.14')

# Outputs: DecimalTuple(sign=1, digits=(3, 1, 4), exponent=-2)
print(my_number.as_tuple())
```

### 4. The Quantize Function for Rounding
Exact Rounding: The .quantize() method rounds a Decimal object to a specific number of decimal places by passing in a Decimal format (like Decimal("1.00") for two places).

Rounding Arguments: You can import specific rounding rules from the decimal module, such as ROUND_HALF_UP, ROUND_HALF_DOWN, and ROUND_HALF_EVEN

```python
from decimal import Decimal, ROUND_HALF_UP, ROUND_HALF_DOWN, ROUND_HALF_EVEN

value = Decimal("10.005")

# Rounding Up
value_up = value.quantize(Decimal("0.00"))
print(f"Round Half Up: {value_up}") # 10.01
```

```python
value = Decimal("10.015")

# Rounding Down
value_down = value.quantize(Decimal("1.00"), ROUND_HALF_DOWN)
print(f"Round Half Down: {value_down}") # 10.00
```

```python
# Rounding Even
value_down = Decimal("10.035").quantize(Decimal("1.00"))
print(f"Round Half Down: {value_down}") # 10.00

value_down = Decimal("10.015").quantize(Decimal("1.00"))
print(f"Round Half Down: {value_down}") # 10.00
```

### 8. Searching Strings with the in Keyword
The in Operator: You can quickly check if a specific substring exists inside a larger string by using the in keyword, which evaluates to a boolean True or False.

Case Sensitivity: Searching with the in keyword is strictly case-sensitive. Searching for "Million" will return False if the string only contains "million".

Whitespace Matters: Spaces are treated as characters. Searching for " million " (with spaces around it) ensures you are finding an exact word, but it will fail if the word is attached to a period or at the end of the sentence.

```python
spam = "Congratulations. You've won a million dollars."

# Basic search
print("million" in spam) # Output: True

# Case-sensitive search
print("Million" in spam) # Output: False

# Checking for exact words with spaces
print(" on " in spam) # Output: False (because it ends with a period, not a space)
```

### Create Account

```python
def get_password():

    while True:
        password = input("Enter password:        ").strip()
        digit = False
        cap_letter = False

        for char in password:
            if char.isdigit():
                digit = True
            elif char.isupper():
                cap_letter = True

        if digit == False or \
           cap_letter == False or \
           len(password) < 8:
            print(f"Password must be 8 characters or more \n"
                  f"with at least one digit and one uppercase "\
                  f"letter.")
        else:
            return password
```

```python
def get_first_name(full_name):
    index1 = full_name.find(" ")
    first_name = full_name[:index1]

    return first_name
```

# Hangman Game

```python
import random

with open("../data/mobydick.txt") as file:
  words = file.readlines()

  words = [sentence.split() for sentence in words]
  print(words)


def get_random_word():
    word = random.choice(words)
    return word
```

```python
# Get a random word from the word list
def get_word():
    word = get_random_word()
    return word.upper()

# Add spaces between letters
def add_spaces(word):
    word_with_spaces = " ".join(word)
    return word_with_spaces

# Draw the display
def draw_screen(num_wrong, num_guesses, guessed_letters, displayed_word):
    print("-" * 79)
    print("Word:", add_spaces(displayed_word),
          " Guesses:", num_guesses,
          " Wrong:", num_wrong,
          " Tried:", add_spaces(guessed_letters))
```

```python
# Get next letter from user
def get_letter(guessed_letters):
    while True:
        guess = input("Enter a letter: ").strip().upper()

        # Make sure the user enters a letter and only one letter
        if guess == "" or len(guess) > 1:
            print("Invalid entry. ",
                  "Please enter one and only one letter.")
            continue
        # Don't let the user try the same letter more than once
        elif guess in guessed_letters:
            print("You already tried that letter.")
            continue
        else:
            return guess
```

```python
# The input/process/draw technique is common in game programming

def play_game():
    word = get_word()

    word_length = len(word)
    remaining_letters = word_length
    displayed_word = "_" * word_length

    num_wrong = 0
    num_guesses = 0
    guessed_letters = ""
    draw_screen(num_wrong, num_guesses, guessed_letters,
                displayed_word)

    print("-" * 79)

    if remaining_letters == 0:
        print(f"Congratulations! You got it in {num_guesses} guesses.")
    else:
        print("Sorry, you lost.")
        print(f"The word was: {word}")
```

```python
def main():
    print("Play the H A N G M A N game")

    again = "y"
    while again.lower() == "y":
        play_game()
        print()
        again = input("Do you want to play again (y/n)?: ")

    print("Bye!")

if __name__ == "__main__":
    main()
```
