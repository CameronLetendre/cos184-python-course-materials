### 1. Order of Precedence
When multiple operators are used in a single statement, Python follows a specific order of precedence:


Exponentiation (**): Right to left.


Multiplication/Division (*, /, //, %): Left to right.


Addition/Subtraction (+, -): Left to right.


Note: You can use parentheses () to override this order and force certain operations to happen first.

```python
# Multiplication happens before addition
result_1 = 3 + 4 * 5
print("3 + 4 * 5 =", result_1)  # Output: 23

# Parentheses force addition first
result_2 = (3 + 4) * 5
print("(3 + 4) * 5 =", result_2) # Output: 35
```

### 2. Compound Assignment Operators
Compound assignment operators allow you to perform an operation and assign the result to the variable in one step.

| Operator | Example         | Equivalent To          |
|----------|------------------|-------------------------|
| +=       | counter += 1     | counter = counter + 1   |
| -=       | counter -= 1     | counter = counter - 1   |
| *=       | price *= 0.8     | price = price * 0.8     |


```python
score_total = 0
score_total += 70  # Adds 70
score_total += 80  # Adds 80 to the previous 70
print(f"Score Total: {score_total}")

price = 100
price *= 0.8       # Multiplies by 0.8 (20% off)
print(f"Discounted Price: {price}")
```

### 3. Working with Strings
Concatenation: You can join strings using the + operator. If you need to join a number to a string, you must first convert the number using the str() function.

F-Strings: A more modern and readable way to embed expressions inside string literals, using {} placeholders.

```python
first_name = "Bob"
last_name = "Smith"
age = 40

# Concatenation (requires explicit conversion for numbers)
print("Name: " + last_name + ", " + first_name + " Age: " + str(age))

# F-String (handles conversion automatically)
print(f"Name: {last_name}, {first_name} Age: {age}")
```

### 4. Escape Sequences

Escape sequences allow you to insert special characters into strings using the backslash \.

* \n: New line

* \t: Tab

* \" or \': Quotation marks inside a string

* \\: Backslash

```python
# Using newlines and tabs
print("Title:\t\tPython Programming\nQuantity:\t5")

# Including quotes in a string
print("Type \"x\" to exit")
# Alternative: Use single quotes to wrap double quotes
print('Type "x" to exit')
```

### 5. The print() Function Options

The print() function accepts multiple arguments separated by commas. It also has two optional keyword arguments:


sep: Defines what character separates the items (Default is a space ' ').


end: Defines what character is printed at the very end (Default is a newline '\n').

```python
# Default behavior
print(1, 2, 3, 4)

# Custom separator
print(1, 2, 3, 4, sep=' | ')

# Custom end character (prevents moving to a new line)
print("Loading", end='...')
print("Done!")
```

> There is no current python argument for adding a prefix to a string such as the end operator does.

> You can still though concat a prefix in a normal print statement by just putting it in the front


```python
value = 22

print(">>>", value)
print(f">>> {value}")
```

### 6. Getting User Input
The input() Function:

Pauses the program and waits for the user to type something.

Always returns a string (str), even if the user types a number.

Converting Input: To perform math on input, you must convert the string to a numeric type using int() or float()

```python
# Getting string input
name = input("Enter your name: ")
print(f"Hello, {name}!")

quantity = input("Enter the quantity: ")
quantity = int(quantity)

price = input("Enter the price: ")
price = float(price)

print(name, "bought", quantity, "at", price, "per unit.")
```

## Example Implementing all of these
A program that takes an item name, quantity, and price then calculates the total adding in shipping and tax costs.

```python
# 1. Get User Input
item_name = input("Enter the item name: ")
quantity = int(input("Enter quantity: "))
quantity = int(quantity)

price = float(input("Enter price per item: "))
price = float(price)

# 2. Arithmetic & Order of Operations
# Parentheses force the addition to happen before multiplication
total_cost = (price + 2.00) * quantity  # Adding a $2 shipping fee per item

# 3. Compound Assignment
# Apply a 10% tax to the existing total variable
total_cost *= 1.10

# 4. Print with Options & Escape Sequences
print("\nGenerating Receipt", end="...")
print(" Done!\n")                        # '\n' adds extra spacing

# 5. F-String Formatting
print(f"Receipt for:\t{item_name}")      # '\t' tabs the output alignment

total_cost = round(total_cost, 2)
print(f"Final Total:\t${total_cost}") # Formats to 2 decimal places
```

### Rounders Rules

```python
# If no argument is put after the number, it will round to the nearest whole number
print(round(20.212093812))   # 3

# 1. If the number is clearly below .5 → rounds down
print(round(3.2))   # 3
print(round(7.1))  # 7
print()

# 2. If the number is clearly above .5 → rounds up
print(round(3.7))  # 4
print(round(8.9))  # 9
print()

# 3. If the number is exactly halfway (.5) → rounds to the nearest even number
print(round(2.5))  # 2  (2 is even)
print(round(3.5))  # 4  (4 is even)
print(round(4.5))  # 4  (4 is even)
print(round(5.5))  # 6  (6 is even)
```
