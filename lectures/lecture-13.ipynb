### 2. Sorting and Case Sensitivity
By default, the .sort() method is case-sensitive, meaning uppercase letters (like "P" in "Pear") are sorted before lowercase letters (like "a" in "apple").

To perform a case-insensitive sort, you can use the key=str.lower argument within the sort method.

```python
numlist = [5, 15, 84, 3, 14, 2, 8, 10, 14, 25]

numlist.sort()
print(f"Sorted: {numlist}")
```

Using Strings

```python
foodlist = ["orange", "apple", "Pear", "banana"]

foodlist.sort()
print(f"Default sort: {foodlist}")
```

Using functions inside

```python
foodlist = ["orange", "apple", "Pear", "banana"]

foodlist.sort(key=str.lower)
print(f"Case-insensitive sort: {foodlist}")
```

```python
def last_character(item):
  print(item[-1:])
  return item[-1:]
```

```python
foodlist.sort(key=last_character)
print(f"{foodlist}")
```

### 4. foodlist.sort() vs. sorted(foodlist)
foodlist.sort(): This is a list method that sorts the list in place. It modifies the original list and does not return a new one.

sorted(foodlist): This is a built-in function that returns a new sorted list. The original list remains unchanged.

Both support the key=str.lower argument for case-insensitive sorting, but using this on a list with mixed types (like strings and integers) will cause a TypeError

```python
numbers = [25, 14, 10, 8, 2]


numbers.sort()
print(f"Modified original: {numbers}")
```

```python
numbers = [25, 14, 10, 8, 2]

new_numbers = sorted(numbers)
print(f"Original list stays same: {numbers}")
print(f"New list: {new_numbers}")
```

### 7. Assignment vs. Shallow Copy vs. Deep Copy
Assignment (=): This does not create a new object. It creates a new reference to the same original list, meaning changes to one will affect the other.

Shallow Copy (list() or .copy()): This creates a new list object, but if that list contains other lists (nested), the inner lists are still shared references.

Deep Copy (copy.deepcopy()): This creates a completely independent copy of the list and all nested objects within it. Changes to a deep-copied list never affect the original.

```python
import copy

original = [1, [2, 3]]

assigned_list = original
assigned_list.append(4)
print(f"Assignment change affects both: {original}")
```

Shallow


```python
original = [1, [2, 3]]

# Changes to the outer list don't affect original,
# but changes to nested lists DO.
shallow_list = copy.copy(original)
shallow_list.append("X")

print(f"Original after shallow change: {shallow_list}")
print(f"Original after shallow change: {original}")
```

```python
original = [1, [2, 3]]
shallow_list = copy.copy(original)

shallow_list[1].append("X")
print(f"shallow_list after shallow change: {shallow_list}")
print(f"Original after shallow change: {original}")
```

Deep

```python
original = [1, [2, 3]]

# Completely independent
deep_list = copy.deepcopy(original)
deep_list[1].append("Y")

print(f"Original stays safe: {original}")
print(f"Deep copy is separate: {deep_list}")
```

### 8. List Slicing
Slicing: You can extract a portion of a list using the syntax list[start:end]. This creates a new list containing elements from the start index up to (but not including) the end index.

Shortcut Syntax:

list[:n]: Gets everything from the beginning up to index n.

list[n:]: Gets everything from index n to the end of the list.

list[:]: Creates a shallow copy of the entire list.

```python
numlist = [5, 15, 84, 3, 14, 2, 8, 10]

slice1 = numlist[1:4]
print(slice1)
```

```python
first_three = numlist[:3]
print(first_three)
```

```python
last_part = numlist[5:]
print(last_part)
```

Concatenation (+): You can combine two or more lists into a single new list using the + operator.

### 10. Transforming and Filtering Lists
map(function, iterable): Applies a specific function to every item in a list (or any iterable) and returns a map object.

filter(function, iterable): Tests every element in a list against a function that returns True or False. It only keeps the items where the result is True.

list(): Because map and filter return "iterator" objects, you must wrap them in the list() function to convert the results back into a readable list format.

map

```python
def square(n):
    return n * n
```

```python
numbers = [1, 2, 3, 4, 5]

squared_numbers = map(square, numbers)
squared_numbers = list(squared_numbers)
print(squared_numbers)
print(numbers)
```

```python
print(type(map(square, numbers)))
print(type(list(map(square, numbers))))
```

filter

```python
def is_even(n):
    return n % 2 == 0
```

```python
numbers = [1, 2, 3, 4, 5]
even_numbers = list(filter(is_even, numbers))
print(even_numbers)
```

### reduce() from functools module

```python
def add_square(total, current):
    return total + (current * current)
```

```python
import functools

numlist = [1, 2, 3, 4, 5, 6]

total = functools.reduce(add_square, numlist)
print(total)
total = functools.reduce(add_square, numlist, 10)
print(total)
total = functools.reduce(add_square, numlist, start=10)
```

```python
def add_words(total, current):
  return total + " " + current
```

```python
from functools import reduce

inventory = ["staff", "hat", "shoes", "bread", "potion"]

# Concatenates: (("Python" + " ") + "is") + " " ...
sentence = reduce(add_words, inventory)
print(f"Inventory: {sentence}")
```

Number Cruncher Program

```python
import random

def get_duplicates(data):
    dups = []
    for i in range(51):
        count = data.count(i)
        if count >= 2:
            dups.append(i)
    return dups

def crunch_numbers(data):
    total = 0
    for number in data:
        total += number

    average = round(total / len(data))
    median_index = len(data) // 2
    median = data[median_index]
    minimum = min(data)
    maximum = max(data)
    dups = get_duplicates(data)

    print("Average =", average,
          "Median =", median,
          "Min =", minimum,
          "Max =", maximum,
          "Dups =", dups)

def main():
    fixed_tuple = (0,5,10,15,20,25,30,35,40,45,50)

    random_list = [0] * 11
    for i in range(len(random_list)):
        random_list[i] = random.randint(0, 50)
    random_list.sort()

    print("TUPLE DATA:", fixed_tuple)
    crunch_numbers(fixed_tuple)
    print()
    print("RANDOM DATA:", random_list)
    crunch_numbers(random_list)

if __name__ == "__main__":
    main()
```
