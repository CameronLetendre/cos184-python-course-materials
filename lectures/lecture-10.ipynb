### 3. Guess the Number Game
This program generates a random number and asks the user to guess it, providing feedback if the guess is "Too high" or "Too low"

````python
import random

LIMIT = 10

# Display the title
def display_title():
    print("Guess the number!\n")

# playing Game logic
def play_game():
    number = random.randint(1, LIMIT)
    print(f"I'm thinking of a number from 1 to {LIMIT}\n")
    count = 1
    while (guess := int(input("Your guess: "))) != number:
        if guess < number:
            print("Too low.")
        elif guess > number:
            print("Too high.")
        count += 1
    print(f"You guessed it in {count} tries.\n")

def main():
    display_title()
    again = "y"
    while again.lower() == "y":
        play_game()
        again = input("Would you like to play again? (y/n): ")
        print()
    print("Bye!")

if __name__ == "__main__":
    main()
````

### 5. Pig Dice Game
The Pig Dice game is a simple game with the following rules:
1. Roll or Hold: For each turn, players choose to roll a die or "hold" to save their points.
2. Roll a 1: If the player rolls a 1, they lose all points for that turn and the turn ends.
3. Hold: If the player holds, the turn's points are added to their total.


Winning: The game ends when a player reaches 20 points.

**Hierarchy chart for the Pig Dice game**

```
main
├── display_rules
└── play_game
    └── take_turn
        ├── roll_die
        └── hold_turn
```


````python
import random

def display_rules():
    print("Let's Play PIG!")
    print("* See how many turns it takes you to get to 20.")
    print("* Turn ends when you hold or roll a 1.")
    print("* If you roll a 1, you lose all points for the turn.")
    print("* If you hold, you save all points for the turn.\n")

def roll_die(turn, score, score_this_turn):
    die = random.randint(1, 6)
    print("Die:", str(die))
    if die == 1:
        score_this_turn = 0
        turn += 1
        print("Turn over. No score.\n")
        turn_over = True
    else:
        score_this_turn += die
        turn_over = False
    return turn, score, score_this_turn, turn_over

def hold_turn(turn, score, score_this_turn):
    print("Score for turn:", score_this_turn)
    score += score_this_turn
    print("Total score:", score, "\n")
    turn_over = True
    game_over = False
    if score >= 20:
        print("You finished in", turn, "turns!")
        game_over = True
        return turn, score, turn_over, game_over
    turn += 1
    return turn, score, turn_over, game_over

def take_turn(turn, score, game_over):
    print("TURN", turn)
    score_this_turn = 0
    turn_over = False
    while not turn_over:
        choice = input("Roll or hold? (r/h): ")
        if choice.lower() == "r":
            turn, score, score_this_turn, turn_over = roll_die(turn, score, score_this_turn)
        elif choice.lower() == "h":
            turn, score, turn_over, game_over = hold_turn(turn, score, score_this_turn)
        else:
            print("Invalid choice. Try again.")
    return turn, score, game_over

def play_game():
    turn = 1
    score = 0
    game_over = False
    while not game_over:
        turn, score, game_over = take_turn(turn, score, game_over)
    print("\nGame over!")

def main():
    display_rules()
    play_game()

if __name__ == "__main__":
    main()
````

### 8. Accessing Items via Indexing
Each item in a list has an index that represents its position. Python supports both positive and negative indexing:
+1


---


Positive indexes start at 0 for the first item and increase from left to right.

Negative indexes start at -1 for the last item and decrease (become more negative) from right to left.

````python
temps = [48.0, 30.5, 20.2, 100.0, 42.0]

# Positive indexing
print("First item (index 0):", temps[0])
print("Last item (index 4):", temps[4])

# Negative indexing
print("First item (index -5):", temps[-5])
print("Last item (index -1):", temps[-1])
````

### 9. Getting and Setting List Items
Getting an Item
You can assign a specific list item to a variable using its index. However, if you attempt to access an index that does not exist, Python will raise an IndexError: index out of range.

Setting (Modifying) an Item
To change the value of an item in a list, use the assignment operator (=) with the specific index.


````python
# Getting items [cite: 72, 73]
temps = [48.0, 30.5, 20.2, 100.0, 42.0]
first_temp = temps[0] # 48.0
print("Original temp at index 3:", temps[3])

# Setting (modifying) an item [cite: 93]
temps[3] = 98.0
print("Modified temp at index 3:", temps[3])

# Handling IndexError [cite: 80]
try:
    bad_access = temps[10] # This will fail
except IndexError as e:
    print(f"Error caught: {e}")
````

````python
inventory = ["staff", "hat", "shoes",
"bread", "potion", ]

stuff = [inventory[0]]

for i in range(0, len(inventory)): # Loop through inventory list
  for j in range(0, len(stuff)): # Loop through current stuff list
    if inventory[i] < stuff[j]: # if current inventory item is less than current stuff list
      stuff.insert(j, inventory[i])
      break

print(stuff)
````

### How to process the items in a list

for loop

````python
scores = [70, 80, 90, 100]
total = 0

for score in scores:
  total += score
print(total)
````

while loop

````python
scores = [70, 80, 90, 100]
total = 0

i = 0
while i < len(scores):
  total += scores[i]
  i += 1
print(total)
````
