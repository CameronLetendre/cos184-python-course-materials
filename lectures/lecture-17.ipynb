### 1. Advanced F-String Format Specifications
- F-strings can include a colon (:) followed by format specifiers to control the alignment, precision, and type of data being displayed.

- Precision and Comma Separators: Using :.2f rounds a float to two decimal places, while adding a comma (,) inserts thousands separators (e.g., {fp_number:,.2f}).

- Field Widths: You can specify a minimum number of characters for a value (e.g., :15) to align columns of data.

- Percentage Formatting: The % specifier automatically multiplies a number by 100 and appends the percent sign (e.g., :.1%).

- Integer Padding: For integers, you can add leading zeros by specifying the width and using the 0 prefix (e.g., :06d)

```python
fp_number = 12345.6789

print(f"{fp_number}")
```

```python
print(f"{fp_number:,}")
```

```python
print(f"{fp_number:.2f}")
```

```python
print(f"{fp_number:15}")
print(f"{fp_number:015}")
```

```python
print(f"{fp_number:>15}")
print(f"{fp_number:<15}")
print(f"{fp_number:^15}")
```

```python
print(f"{fp_number:15,.2f}")
```

```python
# Floating-point number formatting
fp_number = 12345.6789
print(f"{fp_number:.2f}")
print(f"{fp_number:,.2f}")
print(f"{fp_number:15,.2f}")

# Percentage formatting
fp_percent = 0.12345
print(f"{fp_percent:.0%}")
print(f"{fp_percent:.1%}")

# Integer formatting
int_number = 12345
print(f"{int_number:,d}")
print(f"{int_number:6d}")
```

### 2. The Locale Module
The locale module formats data according to specific regional rules, such as currency symbols and decimal separators.

- setlocale(): Configures the program to use a specific region's rules (e.g., 'de_DE.UTF-8' for Germany).

- locale.currency(): Formats a number as a currency string based on the set locale.

- grouping=True: Ensures that thousands separators are applied according to regional standards.

```python
# For Colab, install the locale in the Linux environment first using
!locale-gen de_DE.UTF-8
```

```python
import locale

locale.setlocale(locale.LC_ALL, 'de_DE.UTF-8')

amount = 1234567.8911

print(locale.currency(amount, grouping=False))
print(locale.format_string("%.2f", amount, grouping=True))
```

### The quantize Function for rounding

```python
from decimal import Decimal, ROUND_HALF_UP, ROUND_HALF_DOWN, ROUND_HALF_EVEN

# How to specify the number of decimal places
value = Decimal("10.005")
print(f"Original value: {value}")

default_rounded = value.quantize(Decimal("0.00"))

print(f"{default_rounded}")
```

Rounding up

```python
from decimal import Decimal, ROUND_HALF_UP, ROUND_HALF_DOWN, ROUND_HALF_EVEN

# Example with ROUND_HALF_UP
value_up = Decimal("10.005")

value_up = value_up.quantize(Decimal("1.00"), rounding=ROUND_HALF_UP)

print(f"Quantize with ROUND_HALF_UP (10.005): {value_up}")
```

```python
value_half_up_test = Decimal("10.015")

value_half_up_test = value_half_up_test.quantize(Decimal("1.00"), rounding=ROUND_HALF_UP)
print(f"Quantize with ROUND_HALF_UP (10.015): {value_half_up_test}")
```

### Future Value Program

| Country | Locale String |
| :--- | :--- |
| United States | `'en_US.UTF-8'` |
| Germany | `'de_DE.UTF-8'` |
| France | `'fr_FR.UTF-8'` |
| Switzerland | `'fr_CH.UTF-8'` |

```python
!locale-gen fr_FR.UTF-8
```

```python
import locale
locale.setlocale(locale.LC_ALL, 'fr_FR.UTF-8')
```

```python
def get_future_value(monthly_investment, yearly_interest, years):
    monthly_interest_rate = yearly_interest / 12 / 100
    months = years * 12

    future_value = Decimal("0.00")

    for i in range(months):
        future_value += monthly_investment
        monthly_interest = future_value * monthly_interest_rate
        future_value += monthly_interest

    future_value = future_value.quantize(Decimal("1.00"))
    return future_value
```

```python
import locale
from decimal import Decimal

def main():
    choice = "y"

    while choice.lower() == "y":
        # convert user input to Decimal and int values
        monthly_investment = Decimal(
            input("Enter monthly investment: "))
        yearly_interest = Decimal(
            input("Enter yearly interest rate: "))
        years = int(input("Enter number of years: "))

        future_value = get_future_value(
            monthly_investment, yearly_interest, years)
        print()

        locale.setlocale(locale.LC_ALL, 'fr_FR.UTF-8')

        # Format currency with locale
        formatted_monthly_investment = locale.currency(monthly_investment, grouping=True)
        formatted_future_value = locale.currency(future_value, grouping=True)

        s1 = 20
        s2 = ">10"

        print(f"{'Monthly investment:':{s1}} {formatted_monthly_investment:{s2}}")
        print(f"{'Interest rate:':{s1}} {yearly_interest:{s2}}")
        print(f"{'Years:':{s1}} {years:{s2}}")
        print(f"{'Future value:':{s1}} {formatted_future_value:{s2}}")
        print()

        choice = input("Continue? (y/n): ")
        print()

if __name__ == "__main__":
    main()
```

```python
def get_future_value(monthly_investment, yearly_interest, years):
    monthly_interest_rate = yearly_interest / 12 / 100
    months = years * 12

    future_value = 0.00

    for i in range(months):
        future_value += monthly_investment
        monthly_interest = future_value * monthly_interest_rate
        future_value += monthly_interest

    future_value = round(future_value, 2)
    return future_value
```
