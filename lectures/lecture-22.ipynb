### 1. Extracting Parts of a Datetime Object
You can access individual components of a datetime object using its built-in attributes, such as year, month, day, hour, minute, second, and microsecond

```python
from datetime import datetime

halloween = datetime(1988, 10, 31)
print(halloween)

year = halloween.year
print(year+1)

month = halloween.month
print(month)

day = halloween.day
print(day)
```

### 2. Checking Parts of a Date Object
You can use conditional logic to check specific attributes of a date, like determining if the current month and day match a particular holiday

```python
today = date.today()

if (today.month == 10 and today.day == 31):
    print("Happy Halloween!")
```

### 3. Creating New Dates Based on the Current Date
By getting the current date and performing basic math on its attributes, you can construct a brand-new date object

```python
from datetime import date

today = date.today()
one_year_ago_today = date(today.year - 1, today.month, today.day)

print(f"Today's date: {today}")
print(f"One year ago today: {one_year_ago_today}")
```

```python
from datetime import date

today = date.today()
next_year = today.year*2
one_year_later_today = date(next_year, today.month, today.day)

print(f"Today's date: {today}")
print(f"One year ago today: {one_year_later_today}")
```

### 4. Comparing Date and Datetime Objects
You can use standard comparison operators (>, <, ==) to compare two date or datetime objects to determine chronological order

```python
from datetime import date

today = date.today()
halloween = date(2028, 10, 31)

if today >= halloween:
    print("Halloween 2021 has come and gone.")

elif today <= halloween:
    print("Halloween 2021 is coming soon.")

elif today == halloween:
    print("Happy Halloween 2021!")
```

### 5. Calculating Days Until a Specific Date
When you subtract one date from another, it returns a time span. You can access the .days attribute of that span to find out exactly how many days are between the two dates

```python
from datetime import date

today = date(2026, 12, 12)
halloween = date(today.year, 10, 31)

if today > halloween:
    next_year = today.year + 1
    halloween = date(next_year, 10, 31)

days_until = (halloween - today).days
print(f"{days_until} day(s) until Halloween.")
```

```python
from datetime import datetime
import locale

def get_arrival_date():
    while True:
        date_str = input("Enter arrival date (YYYY-MM-DD): ")
        try:
            arrival_date = datetime.strptime(date_str, "%Y-%m-%d")                 #TODO convert from string to datetime object
        except ValueError:
            print("Invalid date format. Try again.")
            print()
            continue

        now = datetime.now()
        today = datetime(now.year, now.month, now.day)

        if    arrival_date < today:                   #TODO check if arrival date is before today
            print("Arrival date must be today or later. Try again.")
            print()
        else:
            return arrival_date
```

```python
def get_departure_date(arrival_date):
    while True:
        date_str = input("Enter departure date (YYYY-MM-DD): ")
        try:
            departure_date = datetime.strptime(date_str, "%Y-%m-%d")
        except ValueError:
            print("Invalid date format. Try again.")
            print()
            continue

        if departure_date <= arrival_date:
            print("Departure date must be after arrival date. ",
                  "Try again.")
            print()
        else:
            return departure_date
```
