### 1. Splitting Strings

The split() method allows you to divide a single string into a list of multiple substrings.
- If no arguments are provided, it automatically splits the string at each whitespace character.
- You can pass a specific delimiter, such as a slash or a pipe, to split structured data.
- You can also specify an optional number to limit how many splits occur

```python
# Splitting on whitespace
quotation = '''These are  the
times'''
words = quotation.split()
print(words)
```

```python
# Splitting on a delimiter
date = "11/9/1972"
date_parts = date.split("/")
print(date_parts)
```

```python
# Limiting the number of splits
full_name = "GuidoooovonoooRossum"
name_parts = full_name.split("ooo", 1)
print(name_parts)
```

### 2. Concatenating Strings
You can join multiple strings together using the + and += operators.

Alternatively, f-strings provide a very readable way to insert variables directly into a formatted string

```python
first_name = "Eric"
last_name = "Idle"

full_name_plus = last_name + ", " + first_name
print(full_name_plus)
```

```python
name = "Eric"
seperator = ", "
last_name = "Idle"

name += seperator
name += last_name

print(name)
```

```python
name = "Eric"
last_name = "Idle"

full_name_fstring = f"{last_name}, {first_name}"
print(full_name_fstring)
```

### 3. Joining Strings via Sequences
- The join() method takes a sequence (like a list) and combines its elements into a single string
- The string that calls the join() method acts as the separator between the items.
- This method can also be used to quickly insert characters, like spaces, between every letter of an existing string

```python
# Joining list items with a delimiter
address_parts = ["John Doe", "1500 Any Street", "New York", "NY"]
full_address = "<>".join(address_parts)
print(full_address)
```

```python
letters = "HORSE"
spaced_letters = " ".join(letters)
print(spaced_letters)
```

### 4. Importing Datetime Classes
Before working with dates and times in Python, you need to import the required classes from the built-in datetime module.

You can import them one by one on separate lines, or group them together on a single line for cleaner code.

```python
from datetime import date
from datetime import time
from datetime import datetime

from datetime import date, time, datetime
```

### 5. Getting the Current Date and Time
The date and datetime classes have built-in methods to quickly grab the current system date and time.

date.today() returns just the current date (year, month, day).

datetime.now() returns the current date along with the exact time.

```python
# Fetching the current date
invoice_date = date.today()
print(type(invoice_date))
print(invoice_date)
```

```python
# Fetching the current date and time
invoice_timestamp = datetime.now()
print(invoice_timestamp)
```

### 6. Creating Custom Date and Time Objects
You can create specific, custom dates and times using class constructors.

- The date() constructor requires three arguments: year, month, and day.

- The time() constructor accepts optional arguments for hour, minute, second, and microsecond (using a 24-hour clock).

- The datetime() constructor requires the date arguments (year, month, day) and can optionally take the time arguments.

```python
# Creating a date object (Year, Month, Day)
halloween = date(1988, 10, 31)
print(halloween)
```

```python
# Creating a time object (Hour, Minute)
meeting = time(14, 30)
print(meeting)
```

```python
# Creating a datetime object (Year, Month, Day, Hour, Minute)
appointment = datetime(2021, 10, 28, 14, 30, 1, 1)
print(appointment)
```

```python
# Creating a datetime object including seconds
entry_time = datetime(20, 2, 31, 14, 32, 48)
print(entry_time)
```

### Creating a datetime with both a date and a time

```python
import datetime
halloween = datetime.datetime(1988, 10, 31)

# Code that uses format codes to specify the format
print(f"{halloween}")
print(f"{halloween:%m/%d/%Y}")
print(f"{halloween:%H-%M}")
```

```python
halloween = datetime.datetime(1988, 10, 31, 14, 30)

print(f"{halloween:%b %d, %Y %a}")
print(f"{halloween:%B %d, %H:%M}")
print(f"{halloween:%B %d, %I:%M %p}")
```

```python
# Code that formats for locale
print(f"{halloween:%c}")
print(f"{halloween:%x}")

# Code that uses a variable to format a date
date_format = "%Y-%m-%d"
print(f"{halloween:{date_format}}")
```

### Timer Program

```python
from datetime import datetime, time

def main():
    print("The Timer program")
    print()

    input("Press Enter to start...")
    start_time = datetime.now()                        # Get the time right now
    print("Start time:", start_time)
    print()

    # stop timer
    input("Press Enter to stop...")
    stop_time =  datetime.now()                      # Get the time right now
    print("Stop time: ", stop_time)
    print()

    elapsed_time = stop_time - start_time
    days = elapsed_time.days
    minutes =  elapsed_time.seconds // 60                             # Get the amount of minutes passed
    seconds = elapsed_time.seconds % 60
    microseconds = elapsed_time.microseconds

    # calculate hours and minutes
    hours = minutes // 60
    minutes = minutes % 60

    time_object = time(hours, minutes, seconds, microseconds)      # create time object (Hours, minutes, seconds, microseconds)


    # display results
    print("ELAPSED TIME")

    if days > 0:
        print("Days:", days)
    print("Time:", time_object)

if __name__ == "__main__":
    main()
```
