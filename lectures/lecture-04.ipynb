# 1. Converting Data Types

As we learned, input() always returns a string. If you try to do math on a string, Python raises a TypeError. You must use conversion functions to fix this:

int(data): Converts text or floats into a whole integer.

float(data): Converts text or integers into a decimal number.

round(number, digits): Rounds a number to a specific number of decimal places.

```python
# The Problem: This causes a TypeError because "5" is a string
x = 15
y = "5"
# print(x + y)  # Uncommenting this crashes the program

# The Solution: Convert the string to an int
z = x + int(y)
print(f"15 + 5 = {z}")  # Output: 20
```

# 2. Chaining Input Functions

Python developers "chain" functions together to write cleaner code.

Instead of getting the input in one line and converting it in the next, you can wrap the input() function directly inside int() or float().

```python
# The "Long" Way
quantity_str = input("Enter quantity: ")
quantity = int(quantity_str)

# The "Chained" Way (Standard Practice)
# 1. input() runs first and gets the string
# 2. int() runs second and converts it immediately
quantity = int(input("Enter quantity (Chained): "))
price = float(input("Enter price (Chained): "))

print(f"Total: {quantity * price}")
```

# 3. Rounding Numbers

The round() function returns a floating-point number rounded to the specified number of decimal places.

Syntax: round(number, digits)

If digits is omitted, it rounds to the nearest whole number (but still returns a float type usually, or int depending on version context, though the slide implies standard rounding behavior).

```python
miles = 150
gallons = 5.875
mpg = miles / gallons

print(f"Raw MPG:     {mpg}")        # 25.531914...
print(f"Rounded MPG: {round(mpg, 2)}") # 25.53
```

```python
print(round(2.2))
print(round(2.5))
print(round(2.7))
```

### String Comparisons

```python
# 1. Case Sensitivity ("apple" vs "Apple")
print(f"'apple' < 'Apple': {'apple' < 'Apple'}")

# 2. Length/Prefix ("App" vs "Apple")
print(f"'app' < 'Apple':   {'app' < 'Apple'}")

# 3. String Digits ("1" vs "5")
print(f"'1' < '5':         {'1' < '5'}")

# 4. The Tricky One ("10" vs "5")
print(f"'10' < '5':        {'10' < '5'}")
```

```python
def parse_and_compare(word_1, word_2):

    for i in range(1, len(word_1)):
        left = word_1[:i]
        right = word_2[:i]

        if left > right:
            print(f"Left side '{left}' is larger than Right side '{right}'")
            return
        else:
            print(f"Pass: '{left}' is not larger than '{right}'")

    print("Finished: The left side never became larger than the right side.")

print("'banana'")
parse_and_compare("banana", "baNana")
print("\n")

print("Test 2: 'apple'")
parse_and_compare("appl", "apPle")
print("\n")

print("Test 3: 'zoo'")
parse_and_compare("zoo", "zo0")
```
