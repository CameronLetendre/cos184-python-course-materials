## What is an IDE?

An IDE, or Integrated Development Environment, is one application that bundles every tool you need to write, run, and debug code:

- **Text editor** with syntax highlighting and autocomplete.
- **File explorer** for your project's folders and files.
- **Terminal** for running commands like `pip install ...`.
- **Debugger** that lets you pause your code mid-run and inspect variables.
- **Git integration** so you can commit, push, and pull without leaving the editor.
- **Extensions** for Python support, linters, formatters, and more.

VS Code is technically a code editor that becomes an IDE once you add the Python extension. PyCharm, Spyder, and IDLE are other Python IDEs.

## Colab vs an IDE

We've used Google Colab all semester. Colab is a notebook running in the browser on Google's servers. An IDE runs on your own computer, on real files. That difference shows up in several places:

| | Colab | IDE |
|---|---|---|
| **Where code runs** | Google's cloud VM. The session resets when you close the tab. | Your laptop's Python. Files stay until you delete them. |
| **How code is organized** | Cells you can run in any order. They share variables. | A `.py` file that runs top-to-bottom every time. |
| **Installing packages** | Most are pre-installed. Use `!pip install X` per session. | Install once into a virtual environment (`.venv`). |
| **Files** | Upload through the sidebar or mount Google Drive. | Real folders on your disk. The file is the program. |
| **Output** | Plots, tables, and HTML show inline under each cell. | Prints to the terminal. GUI apps like pygame open in their own window. |
| **When things break** | The runtime disconnects and you re-run the cells. | Real error messages with file paths and line numbers. |

Neither is better. Colab is useful for sharing notebooks, running quick experiments, and working on machines without much CPU or RAM. Google provides free CPU, RAM, and GPUs in the cloud, which is helpful for large datasets or training models. An IDE is what you use to build a project that lives on your own machine.

## Quick Review

### Try / Except / Else / Finally
The full exception statement. `else` runs only on success, `finally` runs no matter what.

````python
try:
    number = int(input("Enter an integer: "))
except ValueError as e:
    print(f"Bad input: {e}")
else:
    print(f"You entered {number}.")
finally:
    print("Done.")
````

### Dictionaries of Lists
A dictionary's values can be lists. Use the key to grab the whole list, then index into it like any other list: `dictionary[key][index]`.

````python
students = {
    "Joel": [85, 95, 70],
    "Anne": [95, 100, 100],
    "Mike": [77, 70, 80, 85],
}

print(students["Anne"])
print(students["Anne"][1])

# Average each student's scores
for name, scores in students.items():
    print(f"{name:<6} avg = {sum(scores) / len(scores):.1f}")
````

### The `raise` Keyword
`raise` lets you throw your own exceptions. Useful when something is technically valid Python but wrong for your program - like an empty filename or a negative age. Pair it with `try`/`except` to catch it the same way you'd catch a built-in error.

````python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    return age

try:
    set_age(-5)
except ValueError as e:
    print(f"Caught it: {e}")
````

## The Standard Library - Already Installed

Python ships with hundreds of useful modules. We've already used a few (`csv`, `datetime`, `random`, `sys`). Here are some more worth knowing - **no `pip install` needed**.

### `collections.Counter` - the Word Counter in 3 lines
Remember the Word Counter from Lecture 23? We wrote a whole `for` loop with `if word in counts:`. `Counter` does that for us.

````python
from collections import Counter

text = "the quick brown fox jumps over the lazy dog the fox is quick the dog is lazy"
counts = Counter(text.split())
print(counts.most_common(5))
````

### `statistics` - mean, median, stdev for free

````python
import statistics

scores = [85, 95, 70, 100, 88, 92, 77]
print(f"mean   = {statistics.mean(scores)}")
print(f"median = {statistics.median(scores)}")
print(f"stdev  = {statistics.stdev(scores):.2f}")
````

### `json` - save and load nested data
CSV is great for flat tables. JSON handles nested dicts perfectly - one line to save, one to load.

````python
import json

data = {
    "alice": {"score": 95, "hobbies": ["hiking", "chess"]},
    "bob":   {"score": 88, "hobbies": ["music"]},
}

with open("data.json", "w") as f:
    json.dump(data, f, indent=2)

with open("data.json") as f:
    loaded = json.load(f)

print(loaded)
````

### `pathlib` - the modern way to handle file paths

````python
from pathlib import Path

p = Path("data.json")
print("exists?  ", p.exists())
print("size     ", p.stat().st_size, "bytes")
print("absolute ", p.resolve())
````

## Live Demos - The Fun Part

Each of these is a few lines of code that does something a beginner Python student would have thought was magic on day 1.

### Demo 1: `requests` - one line of internet
GitHub has a free public API. We can ask about a user with no setup, no API key, no auth.

````python
import requests

data = requests.get("https://api.github.com/users/torvalds").json()
print(f"Name:     {data['name']}")
print(f"Company:  {data['company']}")
print(f"Location: {data['location']}")
print(f"Repos:    {data['public_repos']}")
print(f"Bio:      {data['bio']}")
````

### Demo 2: Geolocate any IP address
`ipinfo.io` returns the rough location of any IP. Cybersecurity meets GIS in 4 lines.

````python
import requests

ip = "8.8.8.8"  # Google's public DNS
info = requests.get(f"https://ipinfo.io/{ip}/json").json()
print(f"{ip} -> {info['city']}, {info['region']}, {info['country']}  ({info['loc']})")
````

### Demo 3: A bar chart in 5 lines with `matplotlib`
Real, publication-quality charts.

````python
import matplotlib.pyplot as plt

languages = ["Python", "JavaScript", "Java", "C++", "Go"]
popularity = [29, 19, 16, 11, 7]

plt.bar(languages, popularity)
plt.title("Made-up language popularity")
plt.ylabel("% of devs")
plt.show()
````
