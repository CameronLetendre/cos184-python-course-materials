### 1. Basic Function Definition and Calling
A function is defined using the def keyword. If it doesn't need external data, the parentheses are left empty.

```python
def print_welcome():
    print("Welcome to the Future Value Calculator")
    print("--------------------------------------")
    print()

print_welcome()
```

### 2. Passing Arguments to a Function
Arguments allow you to pass data into a function. Parameters are the variables listed in the function definition, while arguments are the actual values passed during the call.

```python
# A function that accepts one argument
def print_message(message):
    print(message)
    print()

# A function that accepts two arguments and returns a value
def calculate_miles_per_gallon(miles_driven, gallons):
    mpg = miles_driven / gallons
    mpg = round(mpg, 1)
    return mpg

# Calling the functions
miles = 500
gallons = 14.3

print_message("Starting MPG Trip Calculator...")
trip_mpg = calculate_miles_per_gallon(miles, gallons)
print(f"Miles per gallon: {trip_mpg}")
```

```python
# Using multiple function calls inside one line
print(calculate_miles_per_gallon(float(input("How many Miles?")), float(input("How many Gallons?"))))
```

### 3. Local vs. Global Scope
A variable created inside a function is local to that function and cannot be accessed from outside.

```python
def create_msg():
    message = "Hello!"  # Local variable
    return message

my_msg = create_msg()
print(message)
print(my_msg)
```

```python
def create_msg():
    message = "Hello!"  # Local variable
    return message

my_msg = create_msg()
print(my_msg)
```

### 4. Default Values for Arguments
You can assign default values to parameters. If the caller doesn't provide a value, the default is used. Important: Parameters with default values must come after parameters without them.

```python
def calculate_future_value(monthly_investment, yearly_interest, years=20):
    print(f"Calculating for {years} years.")

# Calling with 3 arguments (overrides default)
calculate_future_value(100, 12, 10)

# Calling with 2 arguments (uses default of 20)
calculate_future_value(100, 12)
```

### 5. Keyword Arguments
When calling a function, you can specify the parameter names to pass arguments in a different order than they were defined

```python
def print_billing(name, amount, due_date):
    print(f"Name: {name} | Amount: {amount} | Due: {due_date}")

# Using keyword arguments
print_billing(amount=150.00, due_date="Oct 1st", name="Alex")
```

### 6. The main() Function and Program Flow
In Python, it is best practice to put the "starting code" of your program inside a function named main()

```python
def get_miles():
    miles = float(input("Enter miles: "))
    return miles

def main():
    print("The Miles Driven Program")
    m = get_miles()
    print(f"You entered: {m}")

# This ensures the code only runs if the file is executed directly
if __name__ == "__main__":
    main()
```

### Local vs Global variables

Local Variables

```python
def calc_tax(amount, tax_rate):
    tax = amount * tax_rate  # tax is local variable
    return tax  # return is necessary


def main():
    tax = calc_tax(85.0, .05)  # tax is local variable
    print("Tax:", tax)  # Tax 4.25


if __name__ == "__main__":
    main()
```

Global Variable

```python
tax = 0.0 # tax is global variable

def calc_tax(amount, tax_rate):
    global tax # access global variable
    tax = amount * tax_rate # change global variable

def main():
    calc_tax(85.0, .05)
    print("Tax:", tax) # Tax 4.25 (global)


if __name__ == "__main__":
    main()
```

Global Constant

```python
TAX_RATE = 0.05 # TAX_RATE is global

def calc_tax(amount):
    tax = amount * TAX_RATE # use constant here
    return tax

def main():
    # Example usage:
    item_amount = 100
    total_tax = calc_tax(item_amount)
    print(f"Amount: {item_amount}, Tax Rate: {TAX_RATE}, Calculated Tax: {total_tax}")

if __name__ == "__main__":
    main()
```

### Temperature Module

```python

'''
This module contains functions for converting
between degrees Fahrenheit and degrees Celsius
'''

def to_celsius(fahrenheit):
  '''
  Accepts degrees Fahrenheit (fahrenheit argument)
  Returns degrees Celsius
  '''
  return (fahrenheit - 32) * 5/9

def to_fahrenheit(celsius):
  '''
  Accepts degrees Celsius (celsius argument)
  Returns degrees Fahrenheit
  '''
  return celsius * 9/5 + 32

'''
 the main() function is used to test the other functions
 this code isn't run if this module isn't the main module
'''

def main():
    for temp in range(0, 212, 40):
        print(temp, "Fahrenheit =",
              round(to_celsius(temp)), "Celsius")

    for temp in range(0, 100, 20):
        print(temp, "Celsius =",
              round(to_fahrenheit(temp)), "Fahrenheit")

'''
if this module is the main module, call the main() function
# to test the other functions
'''

if __name__ == "__main__":
    main()
```
