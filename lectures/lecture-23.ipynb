### 1. Creating and Printing Dictionaries
A dictionary in Python stores data in key-value pairs.
* The syntax for creating a dictionary uses curly braces `{}`: `dictionary_name = {key1: value1, key2: value2}`.
* Keys and values can be strings, numbers, or a mix of different data types.
* You can also create an empty dictionary by simply using `{}`.
* Passing the dictionary to the `print()` function will display the entire contents of the dictionary to the console.

```python
countries = {"CA": "Canada", "US": "United States", "MX": "Mexico"}
print(countries)
```

```python
numbers = {1: "One", 2: "Two", 3: "Three", 4: "Four", 5: "Five"}
print(numbers)
```

```python
movie = {"name": "The Holy Grail", "year": 1975, "price": 9.99}
print(movie)
```

### 2. Accessing, Modifying, and Adding Values
* **Accessing:** You can get a value from a dictionary using its key in brackets `dictionary_name[key]`. However, if the key doesn't exist, this throws a `KeyError`.
* **The `in` keyword:** To prevent errors, you can check if a key exists using `key in dictionary` before attempting to access its value.
* **The `get()` method:** Alternatively, use the `get(key, default_value)` method. If the key exists, it returns the value; if not, it safely returns `None` or your specified default value.
* **Modifying/Adding:** You can assign a value to a key using the `=` operator. If the key exists, it updates the value; if the key doesn't exist, it adds a new key/value pair.

```python
countries = {"CA": "Canada", "US": "United States", "GB": "Great Britain", "MX": "Mexico", "US": 1}

print(countries["US"])
```

```python
countries = {"CA": "Canada", "US": "United States", "GB": "Great Britain", "MX": "Mexico"}

print(countries["IE"])
```

```python
countries = {"CA": "Canada", "US": "United States", "GB": "Great Britain", "MX": "Mexico"}

code = "IE"
if code in countries:
    print(countries[code])
else:
    print(f"There is no country for this code: {code}")
```

```python
countries = {"CA": "Canada", "US": "United States", "GB": "Great Britain", "MX": "Mexico"}

# Accessing a value safely using get()
country = countries.get("IE")
print(f"Using get(): {country}")
```

```python
countries = {"CA": "Canada", "US": "United States", "GB": "Great Britain", "MX": "Mexico"}
print(countries)
# Modifying an existing value
countries["GB"] = "United Kingdom"
print(countries)
```

```python
countries = {"CA": "Canada", "US": "United States", "GB": "Great Britain", "MX": "Mexico"}

# Adding a new key/value pair
countries["FR"] = "France"
print(countries)
```

### 3. Deleting Items
There are three primary ways to remove items from a dictionary:
* **`del` keyword:** Use `del dictionary_name[key]` to delete an item. It will throw a `KeyError` if the key doesn't exist.
* **`pop()` method:** Use `pop(key, default_value)`. It removes the item and returns its value. Providing a default value prevents a `KeyError` if the key is missing.
* **`clear()` method:** Use `clear()` to delete all items inside the dictionary at once.

```python
countries = {"CA": "Canada", "US": "United States", "MX": "Mexico"}
print(countries)

# Using the del keyword (checking first to avoid KeyError)
del countries["MX"]
print(countries)
```

Trying del on an invalid key

```python
countries = {"CA": "Canada", "US": "United States", "MX": "Mexico"}
print(countries)

# Using the del keyword (checking first to avoid KeyError)
del countries["IE"]
print(countries)
```

```python
countries = {"CA": "Canada", "US": "United States", "MX": "Mexico"}

# Using the del keyword (checking first to avoid KeyError)
if "CA" in countries:
    del countries["CA"]
else:
  print("Not in Dictionary")
print(countries)
```

### 4. Getting Keys and Values
Dictionaries provide three useful methods to access their contents:
* **`keys()`**: Returns a view of all the keys.
* **`values()`**: Returns a view of all the values.
* **`items()`**: Returns a view of all the key/value pairs as tuples.
You can use a `for` loop alongside these methods to iterate through the data.

```python
countries = {"MX": "Mexico", "US": "United States", "CA": "Canada"}

# Looping through keys
for code in countries.keys():
  countries[code] = "ab"

print(countries)
```

```python
countries = {"MX": "Mexico", "US": "United States", "CA": "Canada"}

# Looping through keys
for code in countries:
    print(code)
```

Values Only

```python
countries = {"MX": "Mexico", "US": "United States", "CA": "Canada"}

# Looping through values
print("\n--- Values ---")
for name in countries.values():
    print(name)
```

Both Keys and Values Simultaneously

```python
# Looping through items
print("\n--- Items ---")
for code, name in countries.items():
    print(f"{code}: {name}")
```

### 5. Constructors and Conversions
Python provides built-in constructors like `list()` and `dict()` for creating and converting data structures.
* **Sorting Keys:** You can extract a dictionary's keys using `.keys()`, convert that view into a list using the `list()` constructor, and then sort it.
* **Converting to a Dictionary:** You can take a two-dimensional list (where each inner list has exactly two elements representing a key and a value) and convert it into a dictionary using the `dict()` constructor.

Dict -> List

```python
# Converting keys to a list and sorting them
countries = {"CA": "Canada", "US": "United States", "MX": "Mexico"}
codes = list(countries.items())
codes.sort()

print(codes, type(codes))
```

List -> Dict

```python
# Converting a two-dimensional list to a dictionary
countries_list = [["GB", "United Kingdom"], ["NL", "Netherlands"], ["DE", "Germany"]]
converted_dict = dict(countries_list)

print(converted_dict)
```

```python
# Converting a three-dimensional list to a dictionary
countries_list = [[[],[]],[[],[]]]
converted_dict = dict(countries_list)

print(converted_dict, type(converted_dict))
```

```python
# Converting a one-dimensional list to a dictionary
countries_list = [1, 2]
converted_dict = dict(countries_list)

print(converted_dict, type(converted_dict))
```

```python
def get_words_from_file(filename):
    with open(filename) as file:
        text = file.read() # read str from file

    text = text.replace("\n", " ") # Replace newlines with space to avoid merging words
    text = text.replace(",", "")
    text = text.replace(".", "")
    text = text.lower()

    words = text.split(" ") # convert str to list
    return [word for word in words if word] # Filter out empty strings that might result from multiple spaces
```

```python
def count_words(words):
    # define a dict to store the word count
    word_count = {}

    for word in words:
        if word in word_count:
            word_count[word] += 1         #TODO increment value count for word key
        else:
            word_count[word] = 1     #TODO add word Key with count value of 1
    return word_count

def display_word_count(word_count):
    words = list(word_count.keys())   #TODO Get only keys and mask to list
    words.sort(key=str.lower)
    for word in words:
        count = word_count[word]
        print(word, "=", count)
```

```python
def main():
    print("The Word Counter program")
    print()

    # change filename to switch text file
    filename = "../data/gettysburg_address.txt"

    # get words, count, and display
    words = get_words_from_file(filename) # get list of words
    word_count = count_words(words) # create dict from list
    display_word_count(word_count)

if __name__ == "__main__":
    main()
```

### Update Operator |=

```python
# A dictionary of an existing book
book = {"title": "1984"}
print(book)

# A dictionary with new data and an updated author name
more_info = {"year": 1949 }
print(f"{more_info}\n")


# Merge the new data into the original book dictionary
book |= more_info

print(book)
print(more_info)
```

```python
# A dictionary of an existing book
book = {"title": "1984", "author": "George Orwell"}

# A dictionary with new data and an updated author name
more_info = {"year": 1949, "author": "G. Orwell"}

# Merge the new data into the original book dictionary
book |= more_info

print(book)
```
