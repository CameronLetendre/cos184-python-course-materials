###1.  String Alignment and Justification
You can align strings in the console using justification methods to create clean, table-like outputs:


ljust(width): Left-aligns the string within a specified width.


rjust(width): Right-aligns the string within a specified width.


Example: print("Hammer".ljust(14), "$9.99".rjust(10)) aligns "Hammer" to the left and the price to the right

```python
print("Hammer".ljust(14), "$9.99")
print("Nails".ljust(14), "$14.50")
print("Screws".ljust(14), "$5.98")
```

```python
print("Hammer".ljust(14), "$9.99".rjust(10))
print("Nails".ljust(14), "$14.50".rjust(10))
print("Screws".ljust(14), "$5.98".rjust(10))
```

###  2. Searching and Modifying Strings
These slides cover methods used to find specific characters or alter the content of a string.


find(str[, start][, end]): Returns the index of the first occurrence of a substring, or -1 if not found.


replace(old, new[, num]): Replaces occurrences of a substring with a new value.


removeprefix(str) and removesuffix(str): Removes specific characters from the beginning or end of a string

```python
# find() example: Finding a character in an email
email = "joel.smith@com"
at_index = email.find("@")
print(at_index)

dot_index = email.find(".", 1)
print(dot_index)

```

```python
# replace() example: Removing dashes from a phone number
phone_number = "555-555-1234"
phone_number = phone_number.replace("-", "o0o")

print(phone_number)
```

```python
# removeprefix/suffix example: Cleaning a string
email = "joel@example.com"
email = email.removeprefix("")
print(email)

email = email.removesuffix(".com")
print(email)
```

###  4. The split() Method
Converts a string into a list of substrings based on a delimiter.Default: Splits on whitespace if no delimiter is provided.Delimiter: You can specify a character (like / or |) to split data.Max Split: A second argument can limit the number of splits performed.

```python
# Default split on whitespace
quotation = '''These
are the     times'''
words = quotation.split()
print(words)
```

```python
# Splitting data records using a delimiter
address = "555o0o555o0o1234"
parts = address.split("o0o")
print(parts)
```

# Hangman Game

```python
import random

with open("../data/function_names.txt", "r") as file:
  words = file.readlines()
```

```python
# Get a random word from the word list
def get_word():
    word = random.choice(words)
    print(word)
    return word

# Add spaces between letters
def add_spaces(word):
    word_with_spaces = " ".join(word)
    return word_with_spaces

# Draw the display
def draw_screen(num_wrong, num_guesses, guessed_letters, displayed_word):
    print("-" * 79)
    print("Word:", add_spaces(displayed_word),
          " Guesses:", num_guesses,
          " Wrong:", num_wrong,
          " Tried:", add_spaces(guessed_letters))
```

```python
# Get next letter from user
def get_letter(guessed_letters):
    while True:
        guess = input("Enter a letter: ").strip()

        if guess == "" or len(guess) != 1:     # How to check to make sure a string is only 1 character
            print("Invalid entry. ",
                  "Please enter one and only one letter.")
            continue

        elif guess in guessed_letters:
            print("You already tried that letter.")
            continue
        else:
            return guess
```

```python
def play_game():
    word = get_word()

    word_length = len(word)
    remaining_letters = word_length
    displayed_word = "_" * word_length


    num_wrong = 0
    num_guesses = 0
    guessed_letters = ""

    while num_guesses < 6:

      draw_screen(num_wrong, num_guesses, guessed_letters,
        displayed_word)

      guess = get_letter(guessed_letters)
      guessed_letters += guess                     # Concatenating a string
      num_guesses += 1


      if guess in word:                     # looking for the guess in the string
        index = word.find(guess)
        displayed_word = displayed_word[:index-1] +  guess + displayed_word[index+1:]     # Switching a specific letter
        remaining_letters -= 1

        if remaining_letters == 0:
          break

      else:
        num_wrong += 1



    print("-" * 79)

    if remaining_letters == 0:
        draw_screen(num_wrong, num_guesses, guessed_letters,
          displayed_word)
        print(f"Congratulations! You got it in {num_guesses} guesses.")
    else:
        print("Sorry, you lost.")
        print(f"The word was: {word}")
```

```python
def main():
    print("Play the H A N G M A N game")

    again = "y"
    while again.lower() == "y":
        play_game()
        print()
        again = input("Do you want to play again (y/n)?: ")

    print("Bye!")

if __name__ == "__main__":
    main()
```
