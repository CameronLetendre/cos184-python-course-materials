### 2. Reading Files with the CSV Module
csv.reader(file): Creates a reader object to iterate through a CSV file. Each row is returned as a list of strings.

next(reader): Advances the reader by one line. This is highly useful for isolating or skipping the header row before looping through the data.

```python
import csv

mode = "r"

with open("../data/people.csv", mode) as file:
    reader = csv.reader(file)
    header = next(reader)

    print(f"Headers: {header}")

    for row in reader:
      print(f"{row}") # Prints the remaining rows
```

```python
import csv

with open("../data/people.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(f"{row}") # Prints the remaining rows
```

### 3. CSV Formatting: Delimiters and Quotes
delimiter: Changes the character used to separate values (e.g., delimiter="\t" creates a tab-separated file).

quotechar / quoting: Determines how quotes are applied to handle cells that might contain the delimiter character naturally.

```python
import csv

tricky_data = [["Apple, Inc.", "Tech"], ["Ford", "Auto"]]

with open("companies.tsv", "w", newline="") as file:
    # Uses tabs instead of commas
    writer = csv.writer(file, delimiter="0")
    writer.writerows(tricky_data)
```

```python
import csv

tricky_data = [["Apple, Inc.", "Tech"], ["Ford", "Auto"]]

with open("companies.tsv", "w", newline="") as file:
    # wrap everythingn in pipes
    writer = csv.writer(file, quotechar='|', quoting=csv.QUOTE_ALL)
    writer.writerows(tricky_data)
```

###4. Opening Multiple Files at Once
You can open and manage multiple files in a single with statement by separating the open() functions with a comma. This is great for reading from one file and writing to another simultaneously

```python
with open("input.csv", "w") as infile: csv.writer(infile).writerows([["apple",1],["pear",2],["orange",3]])
```

```python
import csv

with open("input.csv", "r") as infile, open("output.csv", "a", newline="") as outfile:
    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    for row in reader:
        writer.writerow(row)
```

### 5. The Pickle Module (Binary Serialization)
import pickle: Used to save and load complex Python data structures (like lists and dictionaries) in their native format, rather than converting them to text.
"wb" and "rb": You must use "write binary" and "read binary" modes.
pickle.dump() / pickle.load(): Used to write the object to the file and read it back.

```python
import pickle

my_list = [["Python", 1991], ["Java", 1995]]

# Writing to a binary file
with open("data.bin", "wb") as file:
    pickle.dump(my_list, file)
```

```python
import pickle

# Reading from a binary file
with open("data.bin", "rb") as file:
    recovered_list = pickle.load(file)
    print(recovered_list)
```

### 9. Rounding to Specific Decimal Places
You can combine math.floor or math.ceil with multiplication and division to round up or down to a specific decimal place.

```python
import math

value = 2.0083

# Rounding UP to 1 decimal place
print(math.ceil(value * 10) / 10) # 2.1

# Rounding DOWN to 2 decimal places
print(math.floor(value * 100) / 100) # 2.00
```

### The syntax of an f-string with a format specification

floating-point number: can round a number to a specific number of decimal places

```python
fp_number = 12345.6789

print(f"{fp_number:.1f}")
print(f"{fp_number:,.2f}")
print(f"{fp_number:15,.2f}")

print(f"{fp_number:.2e}")
```

percent: automatically converts a number to a percent and adds a '%' sign

```python
fp_number = 0.12345
print(f"{fp_number:.0%}")
print(f"{fp_number:.1%}")
```

integer: adding integer or adding leading 0's

```python
int_number = 12345
print(f"{int_number:d}")
print(f"{int_number:,d}")
print(f"{int_number:06d}")
```

### Doing accurate calculations with rounding

```python
order_total = float(input("Enter order total: "))

print()

# determine the discount percent
if order_total > 0 and order_total < 100:
    discount_percent = 0
elif order_total >= 100 and order_total < 250:
    discount_percent = .1
elif order_total >= 250:
    discount_percent = .2

# calculate the results
discount = order_total * discount_percent
subtotal = order_total - discount
sales_tax = subtotal * .05
invoice_total = subtotal + sales_tax

# display the results
print(f"Order total: {order_total:10,.2f}")
print(f"Discount amount: {discount:10,.2f}")
print(f"Subtotal: {subtotal:10,.2f}")
print(f"Sales tax: {sales_tax:10,.2f}")
print(f"Invoice total: {invoice_total:10,.2f}")
print()
```

### The quantize Function for rounding

```python
from decimal import Decimal, ROUND_HALF_UP, ROUND_HALF_DOWN, ROUND_HALF_EVEN

# How to specify the number of decimal places
value = Decimal("10.005")
print(f"Original value: {value}")

default_rounded = value.quantize(Decimal("1.00"))
print(f"{default_rounded}")
```

Rounding up

```python
# Example with ROUND_HALF_UP
value_up = Decimal("10.005")

value_up = value_up.quantize(Decimal("1.00"), rounding=ROUND_HALF_UP)

print(f"Quantize with ROUND_HALF_UP (10.005): {value_up}")
```

```python

# Example with another value for ROUND_HALF_UP
value_half_up_test = Decimal("10.015")

value_half_up_test = value_half_up_test.quantize(Decimal("1.0"), rounding=ROUND_HALF_UP)
print(f"Quantize with ROUND_HALF_UP (10.015): {value_half_up_test}")
```

Rounding Down

```python
# Example with ROUND_HALF_DOWN
value_down = Decimal("10.005")

value_down = value_down.quantize(Decimal("1.00"), rounding=ROUND_HALF_DOWN)
print(f"Quantize with ROUND_HALF_DOWN (10.005): {value_down}")
```

Rounding Even

```python

# Example with ROUND_HALF_EVEN
value_half_even_1 = Decimal("10.015")
value_half_even_1 = value_half_even_1.quantize(Decimal("1.00"), rounding=ROUND_HALF_EVEN)
print(f"Quantize with ROUND_HALF_EVEN (10.015): {value_half_even_1}")

value_half_even_2 = Decimal("10.025")
value_half_even_2 = value_half_even_2.quantize(Decimal("1.00"), rounding=ROUND_HALF_EVEN)
print(f"Quantize with ROUND_HALF_EVEN (10.025): {value_half_even_2}")
```
