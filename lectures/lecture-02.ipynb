### 6. Arithmetic Operations

Python supports standard arithmetic operators.

Order of Precedence:


>>>** (Right to left).



>>> *, /, //, % (Left to right).


>>> +, - (Left to right).

```python
# Division vs Integer Division
print("25 / 4 =", 25 / 4)
print("25 // 4 =", 25 // 4)

# Exponentiation
print("3 ** 2 =", 3 ** 2)

# Order of Precedence
# Multiplication happens before addition
result = 3 + 4 * 5
print("3 + 4 * 5 =", result)

# Parentheses change order
result_parens = (3 + 4) * 5
print("(3 + 4) * 5 =", result_parens)
```

# String Operations

### How to Assign strings to variables


```python
first_name = "Bob" # Bob​
last_name = 'Smith' # Smith​
name = "" # (empty string)​
name = "Bob Smith" # Bob Smith

print("first_name", first_name)
print("last_name", last_name)
print("name", name)
```

### Joining Strings

```python
name = last_name + ", " + first_name
print("statement 1", name)

name = f"{last_name}, {first_name}​"
print("statement 2", name)

# name is "Smith, Bob"
```

### How to Join a string and a number

```python
name = "Bob Smith"
age = 40

message = name + " is " + str(age) + " years old." # with the + operator and the str() function
message = f"{name} is {age} years old." # With an f-string​​
```
