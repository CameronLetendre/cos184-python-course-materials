### 1. Relational and Logical Operators
Relational operators are used to compare two values. The result of these comparisons is always a Boolean value: True or False

| Operator | Name |
| --- | --- |
| ==  |  Equal  |
| !=  |  Not Equal  |
| > |  Greater Than  |
| < |   Less Than |
| >= |   Equal to or Greater Than  |
| <= |   Equal to or Less Than |

```python
# Equal
age = 20
print(age == 20)
print(age == 18)

```

```python
# Not Equal
age = 20
print(age != 20)
print(age != 18)
```

```python
# Greater Than
age = 20
print(age < 20)
print(age < 22)
```

```python
# Less Than
age = 20
print(age > 20)
print(age > 18)
```

```python
# Greater Than or Equal to
age = 20
print(age <= 20)
print(age <= 18)
```

```python
# Less Than or Equal to
age = 20
print(age >= 20)
print(age >= 18)
```

### 3. Logical Operators
Logical operators allow you to combine multiple Boolean expressions.

* and: Returns True if both expressions are true.
* or: Returns True if at least one expression is true.
* not: Reverses the Boolean value

```python
year = 2026
quarterback_playing = "Drake Maye"

print(year >= 2026 and quarterback_playing == "Drake Maye")
print()

# Using AND
if year >= 2026 and quarterback_playing == "Drake Maye":
    print("Patriots are going to win the Super Bowl")

# Using OR
if year == 2026 or year == 2018:
    print("Patriots are going to win the Super Bowl")

# Using not
if not quarterback_playing == "Drake Maye":
    print("Uh oh")
```

4. String Comparisons & ASCII Sort Order

When you compare strings using relational operators (<, >, >=), Python does not compare the meaning of the text. Instead, it compares the ASCII/Unicode numeric value of each character, moving from left to right.

```python
# Slide 7 Examples

# 1. Case Sensitivity
print(f"'apple' < 'Apple': {'apple' < 'Apple'}")

# 2. Length/Prefix
print(f"'App' < 'Apple':   {'App' < 'Apple'}")

# 3. String Digits
print(f"'1' < '5':         {'1' < '5'}")

# 4. Comparison by Character
print(f"'10' < '5':        {'10' < '5'}")
```

###5. if-elif-else
* The else clause provides a backup plan. It runs if the if condition is False.


* Use elif (else if) to check multiple conditions in sequence.

Python checks conditions from top to bottom.

As soon as one condition is True, it runs that block and skips the rest.

The else block is optional and runs only if all previous conditions were False.

```python
invoice_total = 200
discount_percent = "Invalid"

if invoice_total >= 500:
  discount_percent = .2
elif invoice_total >= 250:
  discount_percent = .1
elif invoice_total > 0:
  discount_percent = 0
else:
  print("Invoice total must be greater than zero.")

print(f"Discount percent: {discount_percent}")
```

```python
grade = 85

if grade >= 90:
    print("You got an A")
elif grade >= 80:
    print("You got a B")  # This runs because 85 >= 80
elif grade >= 70:
    print("You got a C")  # This is skipped!
else:
    print("You need to study more")
```

```python
score = int(input("Enter test score: "))

if score >= 90 and score <= 100:
    grade = "A"
elif score >= 80 and score < 90:
    grade = "B"
elif score >= 70 and score < 80:
    grade = "C"
elif score >= 60 and score < 70:
    grade = "D"
elif score < 60:
    grade = "F"

print(grade)
```

### Match Statements (Python's Switch)

```python
status_code = int(input("Enter HTTP status code: "))

match status_code:
    case 200:
        print("OK")
    case 400:
        print("Bad Request")
    case 404:
        print("Not Found")
    case 500:
        print("Internal server Error")
    case _:
        print("Unknown Status Code")
```

```python
status_code = int(input("Enter HTTP status code: "))

match status_code:
    case 200:
        print("OK")
    case 400 | 404 | 500:
        print("Bad Request")
    case _:
        print("Unknown Status Code")
```

```python
is_valid = True
customer_type = input("Enter customer type (r/w): ")

if customer_type == "r" or customer_type == "w":
    pass
else:
  print("Customer type must be 'r' or 'w'.")
  is_valid = False

if is_valid:
  invoice_total = float(input("Enter invoice total: "))

  if customer_type.lower() == "r":
      if invoice_total < 100:
          discount_percent = 0
      elif invoice_total >= 100 and invoice_total < 250:
          discount_percent = .1
      elif invoice_total >= 250:
          discount_percent = .2
  elif customer_type.lower() == "w":
      if invoice_total < 500:
          discount_percent = .4
      elif invoice_total >= 500:
          discount_percent = .5
  else:
      discount_percent = 0

  print(f"Discount percent: {discount_percent}")
```

### The Miles Per Gallon program
Using if-statements for error checking

```python
# display a welcome message
print("The Miles Per Gallon program")
print()

# get input from the user
miles_driven = float(input("Enter miles driven: "))
gallons_used = float(input("Enter gallons of gas used: "))

if miles_driven <= 0:
    print("Miles driven must be greater than zero. Try again.")
elif gallons_used <= 0:
    print("Gallons used must be greater than zero. Try again.")
else:
    # calculate and display miles per gallon
    mpg = round((miles_driven / gallons_used), 2)
    print("Miles Per Gallon: ", mpg)

print()
print("Bye!")
```

### The Invoice Program

```python
# display a welcome message

print("The Invoice program")
print()

# get user entries
customer_type = input("Enter customer type (r/w):\t")
invoice_total = float(input("Enter invoice total:\t\t"))
print()

# determine discounts for Retail customers
if customer_type.lower() == "r":
    if invoice_total > 0 and invoice_total < 100:
        discount_percent = 0
    elif invoice_total >= 100 and invoice_total < 250:
        discount_percent = .1
    elif invoice_total >= 250 and invoice_total < 500:
        discount_percent = .2
    elif invoice_total >= 500:
        discount_percent = .25
# determine discounts for Wholesale customers
elif customer_type.lower() == "w":
    if invoice_total > 0 and invoice_total < 500:
        discount_percent = .4
    elif invoice_total >= 500:
        discount_percent = .5
# set discount to zero if neither Retail or Wholesale
else:
    discount_percent = 0

# calculate discount amount and new invoice total
discount_amount = round(invoice_total * discount_percent, 2)
new_invoice_total = invoice_total - discount_amount

# display the results
print(f"Invoice total:\t\t{invoice_total}")
print(f"Discount percent:\t{discount_percent}")
print(f"Discount amount:\t{discount_amount}")
print(f"New invoice total:\t{new_invoice_total}")
print()
print("Bye!")
```
