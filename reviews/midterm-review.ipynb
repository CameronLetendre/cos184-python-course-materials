## 4. Boolean Logic & Operator Precedence

### Logical Operators
| Operator | Meaning |
|----------|---------|
| `and` | Both sides must be `True` |
| `or` | At least one side must be `True` |
| `not` | Flips `True` to `False` and vice versa |

### Precedence (highest to lowest)
```
not  >  and  >  or
```

### String Comparison
Strings compare by Unicode value. Uppercase letters have **lower** values than lowercase (`'A'` = 65, `'a'` = 97).

````python
print(True and False)    # False
print(True or False)     # True
print(not True)          # False
````

Precedence


````python
result = True or False and False

# Step 1: False and False = False
# Step 2: True or False  = True

print(result)   # True
````

````python
# 'not' evaluates first
result = not False and True

# Step 1: not False = True
# Step 2: True and True = True
print(result)   # True
````

````python
# String comparison - Uppercase is always a lower value than Lowercase
print(ord('A'))
print(ord('a'))

print("banana" > "apple")   # True  ('b' > 'a')
print("apple" < "abple")    # True  ('a' > 'A')
````

## 7. while Loops, break & continue

- **`break`** - immediately exits the entire loop
- **`continue`** - skips the rest of the current iteration and goes back to the top

```python
while condition:
    # repeats as long as condition is True
```

````python
# Basic while loop
count = 1

while count <= 5:
    print(count, end=" ")
    count += 1
````

````python
# break: exits the loop entirely when triggered
print()
for i in range(1, 10):
    if i == 5:
        break
    print(i, end=" ")
````

````python
# continue: skips this iteration, keeps going
print()
for i in range(1, 10):
    if i == 5:
        continue
    print(i, end=" ")
````

## 9. Default Parameters & the Walrus Operator

### Default Parameters
A parameter can have a fallback value used when no argument is passed.

**Rules:**
- Parameters with defaults must come **after** parameters without defaults
- `def func(required, optional="default")` is valid
- `def func(optional="default", required)` causes a `SyntaxError`


````python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Sam"))
print(greet("Sam", "Hey"))
````

````python
# defaults must come last
def example(a, b=10, c=20):
    return a + b + c

print(example(5))          # 5 + 10 + 20 = 35
print(example(5, 2))       # 5 + 2  + 20 = 27
print(example(5, 2, 1))    # 5 + 2  + 1  = 8
````

### Walrus Operator `:=`
Assigns a value **and** returns it as part of an expression in one step.

````python
while (n := input("Y/N")).lower() == "y":
    print("Party Time")
````

````python
import random

rolls = []
while (n := random.randint(1, 6)) != 6:
    rolls.append(n)

print("Rolled before getting 6:", rolls)
````


## 10. Scope: Local vs Global Variables

|                   | Local              | Global                 | Global Constant        |
|-------------------|--------------------|------------------------|-------------------------|
| Defined           | Inside a function  | Outside all functions  | Outside all functions   |
| Accessible        | Only inside that function | Anywhere in the file | Anywhere in the file |


````python
# Global variable
total = 100

def show_total():
    local_note = "I only exist inside here"
    print(total)       # can READ global variables
    print(local_note)

show_total()
print(total)
````

````python
# Modifying a global variable requires the 'global' keyword
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
print(counter)   # 2
````

## 11. Lists: Indexing, Slicing & Methods

### Indexing
- Forward index starts at `0`
- Negative index counts from the end: `-1` is last, `-2` is second to last

````python
items = [10, 20, 30, 40, 50]
#         0   1   2   3   4   (forward)
#        -5  -4  -3  -2  -1   (negative)

print(items[0])    # first element
print(items[1])    # second element
print(items[-1])   # last element
print(items[-2])   # second to last
````

### Slicing `list[start:stop]`
- Returns elements from `start` up to (but **not** including) `stop`

````python
items = [10, 20, 30, 40, 50]

# Slicing
print(items[0:-1])
print(items[:3])
print(items[2:])
````

### Key Methods
| Method | What it does |
|--------|--------------|
| `append(item)` | Adds to the end |
| `insert(index, item)` | Inserts before the given index |
| `remove(item)` | Removes first occurrence |
| `pop(index)` | Removes and returns element |
| `sort()` | Sorts in place |

````python
# insert vs append
nums = [1, 2, 3]
nums.append(99)       # adds to END
print(nums)

nums.insert(-1, 50)
print(nums)
````

````python
# insert with a negative index
letters = ["a", "b", "c"]
letters.insert(-1, "z")   # inserts before last element
print(letters)             # ['a', 'b', 'z', 'c']
````

## 14. List Comprehensions

A concise, one-line way to build a list.

```python
[expression for variable in iterable if condition]
```

**Why use them?** Shorter and often more readable than a multi-line for loop.

````python
# loop version
squares = []
for x in range(1, 6):
    squares.append(x ** 2)
print(squares)
````

````python
# comprehension version
squares = [ x for x in range(1, 6) ]
print(squares)
````

with a conditional

````python
evens = [x for x in range(1, 11) if x % 2 == 0]
print(evens)
````

````python
# rewriting a loop with both condition and expression
result = []
for x in range(1, 6):
    if x % 2 == 0:
        result.append(x * 2)
print(result)

# Read this as
# 1. For loop
# 2. conditional
# 3. expression
# 4. add to list

# one-liner version:
result = [x * 2 for x in range(1, 6) if x % 2 == 0]
print(type(result))

````

## 18. Copying Lists

| Method | How | Effect |
|--------|-----|--------|
| Assignment copy | `b = a` | `b` and `a` point to the **same** object changes in one affect both |
| Shallow copy | `b = a.copy()` or `b = a[:]` | New outer list, but nested objects are **still shared** |
| Deep copy | `copy.deepcopy(a)` | Completely independent nothing is shared |

````python
# Assignment copy: both variables point to the SAME list
a = [1, 2, 3]
b = a
b.append(99)
print(a)   # a was changed too
````

````python
# Shallow copy: new list, but nested lists are still shared
a = [1, 2, [10, 20]]
b = a.copy()

b.append(99)       # only affects b
print(a)
print(b)


b[2].append(30)    # changes the nested list
print(a)           # a affected too
print(b)
````

````python
# Deep copy: completely independent
import copy

a = [1, 2, [10, 20]]
b = copy.deepcopy(a)

b[2].append(30)
print(a)   # unaffected
print(b)
````

**Why does it matter?**  
Using an assignment copy when you meant to make a real copy is a common bug. You'll accidentally modify the original. Use `deepcopy` whenever your list contains nested lists.

## 20. Docstrings & Type Hints

### Docstrings
- Written with triple quotes `"""..."""`
- Explains what the function does, its parameters, and return value
- Accessible via `help(function_name)`

### Type Hints
- Annotations showing the **expected type** of parameters and return value
- `param: type` for parameters, `-> type` for return
- Python does **not enforce** them, they are hints for humans and tools

**Why useful?** Make code self-documenting, easier to maintain, and help catch bugs early.

````python
def calculate_area(length: str, width: float) -> float:
    """
    Calculate the area of a rectangle.

    Args:
        length (float): The length of the rectangle.
        width (float): The width of the rectangle.

    Returns:
        float: The area (length * width).
    """
    return length * width

print(calculate_area(5.0, 3.0))   # 15.0
help(calculate_area)
````

## 21. Program Structure & Hierarchy Charts

A **hierarchy chart** shows which functions call which, with `main` at the top.

```
              main
             /      \
        func_a   func_b
                       |
                   func_c
```

````python

def get_number() -> int:
    return int(input("Enter a number: "))

def double(n: int) -> int:
    return n * 2

def display(result: int) -> None:
    print(f"Double is: {result}")
    answer = get_answer() # Function 4

    return answer

def get_answer():
  return input("Do you want to do another number?")

def main():
  reply = 'y'

  while reply == 'y':
    n = get_number() # Function 1

    result = double(n) # Function 2
    reply = display(result) # Function 3

if __name__ == "__main__":
  main()

# Hierarchy:

# main() -> get_number()
#        -> double()
#        -> display() -> get_answer()
````

## 22. File Handling & the `with` Keyword

### Why use `with`?
1. **Automatically closes** the file when the block ends even if an error occurs
2. **Prevents leaks** unclosed files waste memory
3. **Cleaner** no need to call `file.close()` manually

### File modes
| Mode | Meaning |
|------|---------|
| `"r"` | Read |
| `"w"` | Write (overwrites) |
| `"a"` | Append |

> **Intentional errors.** The next two cells both divide by zero on purpose. Run the first, then the second, and compare: which one leaves the file properly closed? That difference is the reason to use `with`.

````python
file = open("review_test.txt", "w")
file.write("asas")
10 /0
````

````python
# With 'with' always closes automatically
with open("review_test.txt", "w") as file:
    file.write("Line one\n")
    10/0
    file.write("Line two\n")

with open("review_test.txt", "r") as file:
    contents = file.read()

print(contents)
````
